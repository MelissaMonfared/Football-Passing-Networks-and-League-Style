<div style="color:#E8694A;font-size:11.5px;letter-spacing:3px;text-transform:uppercase;
margin-bottom:14px;font-weight:bold;">Abstract</div>
<p style="margin:0 0 13px;">Football passing networks are usually studied one team at a
time, and the resulting claims about style are rarely tested against the variation that
ordinary match-to-match noise would produce anyway. This project treats the passing
network as a random object with a nested structure and asks where its variation actually
sits.</p>

<p style="margin:0 0 13px;">The main sample is the only complete balanced cross section in
the StatsBomb open archive, which is the 2015/16 season of La Liga, the Premier League, Serie A and
Ligue 1, giving 1,517 matches, 80 clubs each playing 37 or 38 games, and 3,034 team-match
passing networks built from 5.7 million events. A balanced nested design of this size
allows the variance in network topology to be split cleanly between league, club and
match.</p>

<p style="margin:0 0 13px;">The headline result is that national leagues barely exist as
structural entities. Averaged across 22 network and spatial features, league accounts for
2.2% of variance, club for 23.4%, and the individual match for 74.3%. Club is significant
for every feature tested; league for eight of 22, and those eight are spatial rather than
topological. Leagues differ in where teams play, with final-third share at 12.7% and mean
pass origin at 10.6%, and hardly at all in how their passing graphs are wired.</p>

<p style="margin:0 0 13px;">A second sample provides an embedded natural experiment. The FA
Women's Super League played its entire 2020/21 season behind closed doors, preceded by two
seasons with crowds and followed by a further season with crowds restored, which gives a
treated period bracketed on both sides rather than only interrupted. Across 457 matches and
914 networks the spectator effect is estimated with permutation inference at the level of
the match, alongside a power analysis so that a null result carries a size.</p>

<p style="margin:0;">Throughout, generation of results is separated from interpretation of
them. Predictive models are reported apart from the causal and descriptive analyses, network
construction choices are re-run under alternative specifications, and three data-integrity
failures caught during the build are documented in place rather than quietly repaired.</p>
</div>

<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Introduction</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">Why a passing network needs a variance decomposition before it needs an interpretation</div></div>


A passing network is an appealing object. Draw the players at their average positions, join
them with arrows weighted by how often one passed to the other, and the picture looks like
it explains something. Analysts use these diagrams to describe a team's identity, and the
research literature has built a substantial body of metrics on top of them, among them density,
centralization, entropy, clustering, and a dozen more.

The difficulty is that almost none of this work establishes a baseline. A network drawn from
one match is a single draw from a distribution, and the same team playing the same opponent
a week later will produce a visibly different graph. Without knowing how much a network
varies for reasons that have nothing to do with identity, any claim that a particular
topology reflects a particular style is untestable. The diagram always looks meaningful.

This project starts from the opposite end. Before asking what network structure means, it
asks how network structure varies, and at which level. If the answer is that most variation
sits at the match level, then a single-match network says little about a team, and the
metrics computed on it are mostly measuring the match rather than the side that played it.


<div style="background:#EAF5F3;border-left:4px solid #2AA396;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#2AA396;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Interpretation</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">The question this makes it possible to ask</h4>
Once variance is partitioned, a sharper question becomes available. Football is organised
into national leagues that are routinely described as having distinct characters, with the
Premier League as fast and direct, La Liga as technical, Serie A as tactical. If those
descriptions correspond to anything structural, league should account for a meaningful
share of the variation in how passing networks are built.
<br><br>
That is a testable proposition rather than a stylistic one, and it requires exactly the
kind of sample this project uses. Four leagues, complete seasons, every club playing the
same number of matches, and networks constructed identically throughout. A balanced nested
design of that shape gives the variance components in closed form, without relying on an
optimiser to converge.
</div>


**What the second sample adds.** Variance decomposition describes structure but cannot
speak to cause. The FA Women's Super League provides an unusually clean intervention. The
entire 2020/21 season was played behind closed doors, with two full seasons of crowd
football before it and another after. A treated period bracketed on both sides is stronger
than one that is merely interrupted, because a return to baseline is itself evidence that
the change tracked the intervention rather than a trend.

The sample is small and the archive has a two-season gap, so this part of the project is
framed as an embedded case study with its uncertainty stated rather than as a definitive
causal estimate. Section 19 reports what effect size the design could have detected, so
that a null carries information instead of ambiguity.

<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Research Questions</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">Six questions, each tied to a section and a decision rule</div></div>


**RQ1.** How is the variance in passing network topology distributed across league, club and
match, and does any structural feature show a league-level signature once club is accounted
for?

**RQ2.** Do the features that distinguish clubs describe the same aspects of play as those
that distinguish leagues, or do the two levels separate along different dimensions?

**RQ3.** Does the spatial organisation of passing, meaning where on the pitch play happens
and how widely it is distributed, behave differently from the topological organisation,
meaning how the passing graph is wired?

**RQ4.** Are network metrics associated with match outcomes once team quality is controlled
for, and does that association survive the alternative network definitions of Section 14?

**RQ5.** Did the removal of spectators change the structure of team passing, and if the
answer is no, what size of effect can the design rule out?

**RQ6.** Do network features add predictive information about match results beyond what
team identity and basic match statistics already provide?


<div style="background:#EEF3F5;border-left:4px solid #22414E;border-radius:3px;

color:#2F4750;font-size:14px;"><div style="color:#22414E;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Method</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">Rules fixed before any model was fitted</h4>
The primary analysis is the variance decomposition of RQ1 on the four-league cross section,
which is confirmatory. Everything concerning player roles, spatial structure, nonlinear
relationships and prediction is secondary and labelled as such where it appears.
<br><br>
Three standing rules apply. Uncertainty for anything measured at team-match level is
estimated by resampling matches, never individual passes, because passes within a match are
strongly dependent. Any comparison spanning more than a handful of features carries a
Benjamini-Hochberg adjustment with the number of tests reported. And a conclusion is called
robust only if it survives the alternative network constructions in Section 14, which vary
whether set pieces are included, whether the graph is directed, and where the edge weight
threshold sits.
<br><br>
The spectator analysis of RQ5 is treated as a case study rather than a headline causal
estimate, for the reasons set out there. The treated period is one season, the archive
omits the two seasons that followed it, and the season immediately before it was abandoned
partway through.
</div>

<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Data and Provenance</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">What the archive contains, and why this sample rather than another</div></div>


All data come from StatsBomb Open Data, an event-level archive released for public research
use. Every touch of the ball is recorded with a type, a location on a 120 by 80 pitch, the
player and team responsible, and for passes the recipient and end location. That recipient
field is what makes passing networks possible; match-level feeds do not carry it.

The archive is not a sample of world football. It is a set of releases chosen by the
provider, so what is available constrains the design rather than the other way round. That
constraint was resolved before any analysis by auditing the whole archive rather than
assuming a desired competition would be present.


```python
DATA = '/kaggle/working/fb_data'   # see Section 25 for how this folder is built

# Kaggle notebooks run with the internet switched off by default, which stops both
# pip and any live API call. The competitions index is therefore read from the cached
# copy shipped with the data, and statsbombpy is used only if it happens to be present
# and reachable. Nothing else in the notebook needs a network connection, because every
# event file was downloaded once and cached as parquet.
def load_competitions():
    cached = f'{DATA}/competitions.parquet'
    try:
        from statsbombpy import sb as _sb
        comps = _sb.competitions()
        print("competitions index read live from StatsBomb")
        return comps
    except Exception as exc:
        if os.path.exists(cached):
            print(f"live access unavailable ({type(exc).__name__}), "
                  f"using the cached competitions index")
            return pd.read_parquet(cached)
        raise RuntimeError(
            "No competitions index. Either enable internet for this notebook, "
            "or attach the cached data folder described in Section 25.") from exc


competitions = load_competitions()
club_comps = competitions[~competitions.competition_international.fillna(False)]

coverage = (club_comps.groupby(['country_name', 'competition_name', 'competition_gender'])
            .agg(seasons=('season_id', 'size'),
                 season_list=('season_name', lambda s: ', '.join(sorted(s))))
            .sort_values('seasons', ascending=False))
print(f"Competitions in the open archive: {len(competitions)} competition-seasons, "
      f"{competitions.competition_name.nunique()} distinct competitions")
coverage.head(10)
```

<div style="background:#FBEFE8;border-left:4px solid #C1613C;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#C1613C;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Caution</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">The design was chosen by the archive, not by preference</h4>
A natural first instinct for a spectator-absence study is the Bundesliga 2019/20 restart,
which is the most discussed closed-door period in European football. The open archive holds
Bundesliga for 2015/16 and 2023/24 only. There is no 2019/20, so that study cannot be run
with event data no matter how it is specified.
<br><br>
La Liga looks more promising with 18 seasons, until the coverage is examined season by
season, because every season except 2015/16 contains only Barcelona's matches, since those
releases accompanied a player-specific data project. A 33-match season is not a league.
<br><br>
What the archive does support is set out below, and it is two samples rather than one. The
point of running this check first is that the alternative is discovering the gap after
building an analysis around a competition that is not there.
</div>

```python
CROSS_SECTION = {'La Liga': (11, 27), 'Premier League': (2, 27),
                 'Serie A': (12, 27), 'Ligue 1': (7, 27)}
PANEL = {'FAWSL 2018/19': (37, 4), 'FAWSL 2019/20': (37, 42),
         'FAWSL 2020/21': (37, 90), 'FAWSL 2023/24': (37, 281)}

MATCH_TABLES = {'cross': f'{DATA}/matches_cross.parquet',
                'panel': f'{DATA}/matches_panel.parquet'}

def season_summary(spec, table):
    """Season shape, read from the cached match index so no network call is needed."""
    meta = pd.read_parquet(MATCH_TABLES[table])
    rows = []
    for name, (cid, sid) in spec.items():
        m = meta[(meta.competition_id == cid) & (meta.season_id == sid)]
        teams = set(m.home_team) | set(m.away_team)
        per_team = pd.concat([m.home_team, m.away_team]).value_counts()
        rows.append({'sample': name, 'matches': len(m), 'clubs': len(teams),
                     'matches per club': f"{per_team.min()} to {per_team.max()}",
                     'complete round robin': len(m) >= 0.95 * len(teams) * (len(teams) - 1),
                     'first': str(m.match_date.min())[:10],
                     'last': str(m.match_date.max())[:10]})
    return pd.DataFrame(rows)

cross_summary = season_summary(CROSS_SECTION, 'cross')
panel_summary = season_summary(PANEL, 'panel')
print("Cross section: the only complete balanced seasons in the archive")
print(cross_summary.to_string(index=False))
print(f"\n  total {cross_summary.matches.sum()} matches, "
      f"{cross_summary.clubs.sum()} club-seasons, "
      f"{2*cross_summary.matches.sum()} team-match networks\n")
print("Panel: the spectator-absence case study")
print(panel_summary.to_string(index=False))
print(f"\n  total {panel_summary.matches.sum()} matches")
```

<h3 style="color:#22414E;font-family:Helvetica,Arial,sans-serif;font-size:17px;
border-bottom:1.6px solid #CFD8DC;padding-bottom:7px;margin:26px 0 12px;">
<span style="color:#E8694A;">5.1</span>&nbsp;&nbsp;External validation against the historical record</h3>


The checks above are internal. They confirm the pipeline is consistent with itself, which is
necessary and not sufficient, since a systematic error applied uniformly would pass all six.

The stronger test is external. Every goal in this corpus is reconstructed from individual shot
and own-goal events, never read from a results feed. Aggregating those reconstructed goals into
points across a season produces a league table, and that table can be compared against the
official historical record. If the event parsing, the own-goal handling, the club identity
resolution and the home and away assignment are all correct, the tables will match exactly. If
any one of them is wrong, they will not.


```python
tm = team_match.copy()
tm['pts'] = np.where(tm.goals > tm.goals_against, 3,
                     np.where(tm.goals == tm.goals_against, 1, 0))

def league_table(league):
    t = (tm[tm.league == league].groupby('team')
         .agg(P=('pts', 'size'),
              W=('pts', lambda s: int((s == 3).sum())),
              D=('pts', lambda s: int((s == 1).sum())),
              L=('pts', lambda s: int((s == 0).sum())),
              GF=('goals', 'sum'), GA=('goals_against', 'sum'),
              Pts=('pts', 'sum')))
    t['GD'] = t.GF - t.GA
    return t.sort_values(['Pts', 'GD'], ascending=False)[
        ['P', 'W', 'D', 'L', 'GF', 'GA', 'GD', 'Pts']]

# The official 2015/16 finishing positions, taken from the public record and hard-coded
# here purely as a comparison target. Nothing downstream uses them.
OFFICIAL = {
    'Premier League': [('Leicester City', 38, 23, 12, 3, 68, 36, 81),
                       ('Arsenal', 38, 20, 11, 7, 65, 36, 71),
                       ('Tottenham Hotspur', 38, 19, 13, 6, 69, 35, 70)],
    'La Liga': [('Barcelona', 38, 29, 4, 5, 112, 29, 91),
                ('Real Madrid', 38, 28, 6, 4, 110, 34, 90),
                ('Atlético Madrid', 38, 28, 4, 6, 63, 18, 88)],
    'Serie A': [('Juventus', 38, 29, 4, 5, 75, 20, 91),
                ('Napoli', 38, 25, 7, 6, 80, 32, 82),
                ('AS Roma', 38, 23, 11, 4, 83, 41, 80)],
}

rows = []
for lg, official in OFFICIAL.items():
    rec = league_table(lg)
    for name, P, W, D, L, GF, GA, Pts in official:
        if name not in rec.index:
            rows.append({'league': lg, 'club': name, 'match': 'club not found'})
            continue
        r = rec.loc[name]
        exact = (int(r.P) == P and int(r.W) == W and int(r.D) == D and int(r.L) == L
                 and int(r.GF) == GF and int(r.GA) == GA and int(r.Pts) == Pts)
        rows.append({
            'league': lg, 'club': name,
            'reconstructed': f"{int(r.W)}-{int(r.D)}-{int(r.L)}  {int(r.GF)}:{int(r.GA)}  "
                             f"{int(r.Pts)} pts",
            'official': f"{W}-{D}-{L}  {GF}:{GA}  {Pts} pts",
            'exact match': exact})
validation = pd.DataFrame(rows)
print(f"Clubs checked against the official record: {len(validation)}")
print(f"Exact matches on played, won, drawn, lost, scored, conceded and points: "
      f"{int(validation['exact match'].sum())} of {len(validation)}")
validation
```

color:#2F4750;font-size:14px;"><div style="color:#22414E;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Method</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">Football conventions used throughout, and why each is standard</h4>
<strong>Pitch coordinates.</strong> StatsBomb records positions on a 120 by 80 grid with the
attacking direction always left to right, so both teams are oriented the same way regardless of
which end they started at. The observed ranges in this corpus are 0.7 to 120.0 and 0.1 to 80.0,
consistent with that specification.
<br><br>
<strong>Thirds of the pitch.</strong> The defensive third is x below 40, the middle third 40 to
80, and the final third 80 and above, which divides the 120-unit pitch into three equal parts.
This is the standard analytical convention and it is what the final-third share measures.
<br><br>
<strong>Set pieces.</strong> Corners, free kicks, throw-ins, goal kicks and kick-offs are
restarts from a dead ball. They are excluded from the default network because the recipient is
determined by a rehearsed routine rather than by open-play structure, and they are re-included
in the Section 14 sensitivity check.
<br><br>
<strong>Points.</strong> Three for a win and one for a draw, the system in use in all four
leagues throughout 2015/16.
<br><br>
<strong>Own goals.</strong> Credited to the team that benefits, not to the player who scored
them, which is how league tables are compiled and why Section 5 had to correct for it.
</div>

<div id="sec6" style="border-left:5px solid #E8694A;padding:16px 0 16px 20px;
margin:30px 0 14px;font-family:Helvetica,Arial,sans-serif;">
<span style="color:#E8694A;font-size:12.5px;letter-spacing:2.6px;font-weight:bold;">SECTION 6</span>
<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Constructing a Passing Network</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">Every decision changes the graph, so every decision is stated</div></div>


A passing network for one team in one match is a directed weighted graph. Nodes are players,
and an edge from player $i$ to player $j$ carries the number of completed passes between
them in that direction.


<div style="background:#FCFAF5;border:1px solid #CFD8DC;border-radius:3px;
padding:16px 20px;margin:14px 0;text-align:center;">$$G_{tm} = (V, E), \qquad
V = \{\text{players of team } t \text{ appearing in match } m\}, \qquad
w_{ij} = \big|\{\text{completed passes } i \rightarrow j\}\big|$$</div>

<div style="background:#EEF3F5;border-left:4px solid #22414E;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#22414E;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Method</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">Four construction decisions, and what each one costs</h4>
<strong>Completed passes only.</strong> An incomplete pass has no recipient and therefore
cannot define an edge. Incomplete passes are retained separately as a volume measure, so the
information is not lost, but they do not enter the topology.
<br><br>
<strong>Set pieces excluded by default.</strong> A corner or free kick has a recipient chosen
by a rehearsed routine rather than by the open-play structure the network is meant to
describe. Including them adds edges that reflect the set-piece coach. Section 14 re-runs the
main results with them included.
<br><br>
<strong>Substitutes included as nodes.</strong> Restricting to starters would make the node
count depend on how many changes a manager made, which is a coaching decision rather than a
property of the passing structure. The cost is that node count varies, between 11 and 14 in
the cross section and 11 and 16 in the panel, which is why every centralization measure below
is normalised by node count.
<br><br>
<strong>Edge weights unthresholded by default.</strong> A single pass between two players is
a real edge. Thresholding removes noise but also removes genuine low-frequency connections,
and the threshold is arbitrary. Section 14 tests a minimum of three passes.
</div>

```python
SET_PIECE_TYPES = {'Corner', 'Free Kick', 'Throw-in', 'Kick Off', 'Goal Kick'}

def passes_for_team(events, team, include_set_pieces=False, completed_only=True):
    """Pass events defining one team's network in one match."""
    p = events[(events.type == 'Pass') & (events.team == team)].copy()
    if completed_only:
        p = p[p.pass_outcome.isna()]      # StatsBomb leaves this null when complete
    if not include_set_pieces and 'pass_type' in p.columns:
        p = p[~p.pass_type.isin(SET_PIECE_TYPES)]
    return p[p.player.notna() & p.pass_recipient.notna()]


def build_graph(passes, directed=True, weighted=True, min_weight=1):
    edges = passes.groupby(['player', 'pass_recipient']).size().reset_index(name='w')
    edges = edges[edges.w >= min_weight]
    G = nx.DiGraph() if directed else nx.Graph()
    G.add_nodes_from(set(passes.player) | set(passes.pass_recipient))
    for _, r in edges.iterrows():
        w = int(r.w) if weighted else 1
        if G.has_edge(r.player, r.pass_recipient):
            G[r.player][r.pass_recipient]['weight'] += w
        else:
            G.add_edge(r.player, r.pass_recipient, weight=w)
    return G


# one worked example, so the object is concrete before it is aggregated
example_id = int(team_match.total_passes.idxmax())
example = team_match.loc[example_id]
ev = pd.read_parquet(f"{DATA}/events/{int(example.match_id)}.parquet")
p_ex = passes_for_team(ev, example.team)
G_ex = build_graph(p_ex)
print(f"{example.team} against {example.opponent}, "
      f"{example.league}, matchweek {example.match_week}")
print(f"  completed open-play passes : {len(p_ex)}")
print(f"  nodes (players involved)   : {G_ex.number_of_nodes()}")
print(f"  edges (distinct pairings)  : {G_ex.number_of_edges()}")
print(f"  density                    : {nx.density(G_ex):.3f}")
print(f"  heaviest pairing           : "
      f"{max(G_ex.edges(data=True), key=lambda e: e[2]['weight'])[:2]} "
      f"({max(d['weight'] for _,_,d in G_ex.edges(data=True))} passes)")
```

<div id="sec7" style="border-left:5px solid #E8694A;padding:16px 0 16px 20px;
margin:30px 0 14px;font-family:Helvetica,Arial,sans-serif;">
<span style="color:#E8694A;font-size:12.5px;letter-spacing:2.6px;font-weight:bold;">SECTION 7</span>
<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Network Metrics and Their Meaning</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">What each quantity measures, and what it cannot</div></div>


Twenty-two features are computed for every team-match network, in four families. Each is
defined here with the football question it speaks to, because a metric with no interpretation
attached is a number in a table rather than a measurement.


<div style="background:#FCFAF5;border:1px solid #CFD8DC;border-radius:3px;
padding:16px 20px;margin:14px 0;text-align:center;">$$\text{density } D = \frac{|E|}{|V|(|V|-1)}
\qquad
\text{strength } s_i = \sum_j \big(w_{ij} + w_{ji}\big)
\qquad
\text{entropy } H = -\sum_k p_k \log p_k$$</div>

<div style="background:#FCFAF5;border:1px solid #CFD8DC;border-radius:3px;
padding:16px 20px;margin:14px 0;text-align:center;">$$\text{weighted reciprocity } R = \frac{\sum_{ij}\min(w_{ij}, w_{ji})}{\sum_{ij} w_{ij}}
\qquad
\text{Gini } G = \frac{2\sum_i i\,s_{(i)}}{n\sum_i s_i} - \frac{n+1}{n}$$</div>

<div style="background:#FCFAF5;border:1px solid #CFD8DC;border-radius:3px;
padding:16px 20px;margin:14px 0;text-align:center;">$$\text{Freeman centralization } C = \frac{\sum_i \big(c_{\max} - c_i\big)}{(n-1)(n-2)}$$</div>


Density asks how many of the possible passing relationships a team actually used. A side that
circulates through every pairing scores high; one that plays down fixed channels scores low.

Weighted reciprocity asks how much passing goes both ways. A high value means players
exchange the ball with each other rather than moving it in one direction through the side,
which distinguishes circulation from progression.

Entropy over edge weights measures how evenly passing volume is spread across pairings, and
entropy over player strength does the same across individuals. Gini on strength captures the
same idea as inequality rather than spread, and the two are reported together because they
disagree in informative ways when one player dominates a network that is otherwise even.

Freeman centralization measures how far a network sits from a perfect star, normalised by the
maximum achievable for that node count. This matters because node counts vary with
substitutions, from 11 to 14 in the cross section and 11 to 16 in the panel, and an
unnormalised centralization would confound structure with how many changes a manager made.


<div style="background:#FBEFE8;border-left:4px solid #C1613C;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#C1613C;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Caution</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">One implementation detail that is easy to get backwards</h4>
Betweenness centrality treats edge weights as distances, so a heavily used pairing would count
as a <em>long</em> path under the default. That is the opposite of what passing volume means.
Weights are inverted before computing betweenness, so a pairing used 26 times is a short hop
and a pairing used once is a long one.
<br><br>
Separately, eigenvector centrality via the numpy solver refuses directed graphs that are not
strongly connected, and a passing network usually is not, because a substitute who receives a late
pass and never completes one is a sink. That condition occurs in 18% of networks in this
corpus. The power-iteration solver handles it, and the fallback is the undirected projection.
Left unhandled, an eighth of the corpus would have dropped out of every eigenvector analysis,
and the missingness would have correlated with substitution patterns rather than being random.
</div>

<div id="sec9" style="border-left:5px solid #E8694A;padding:16px 0 16px 20px;

<span style="color:#E8694A;font-size:12.5px;letter-spacing:2.6px;font-weight:bold;">SECTION 9</span>
<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Where the Variance Lives</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">The primary analysis, and the result the rest of the project is organised around</div></div>


A passing network varies for three kinds of reason. Some variation is between leagues, if
national competitions really do have structural characters. Some is between clubs within a
league, which is what most people mean by a team's style. And some is between matches played
by the same club, which is opponent, scoreline, personnel and chance.

Partitioning those three is the question this sample was chosen to answer, and it requires a
balanced nested design, with club nested in league and matches nested in club.


<div style="background:#FCFAF5;border:1px solid #CFD8DC;border-radius:3px;
padding:16px 20px;margin:14px 0;text-align:center;">$$y_{ijk} = \mu + \alpha_i + \beta_{j(i)} + \varepsilon_{ijk},
\qquad
\alpha_i \sim N(0,\sigma^2_{\text{league}}), \quad
\beta_{j(i)} \sim N(0,\sigma^2_{\text{club}}), \quad
\varepsilon_{ijk} \sim N(0,\sigma^2_{\text{match}})$$</div>

<div style="background:#EEF3F5;border-left:4px solid #22414E;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#22414E;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Method</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">Why the components are computed in closed form rather than fitted</h4>
With 20 clubs in every league and 37 or 38 matches for every club, the design is balanced, and
the expected mean squares of a nested random-effects ANOVA give the variance components
exactly:
<br><br>
$E[MS_{\text{league}}] = \sigma^2_{\text{match}} + n\sigma^2_{\text{club}} + nc\,\sigma^2_{\text{league}}$,
$\;E[MS_{\text{club}}] = \sigma^2_{\text{match}} + n\sigma^2_{\text{club}}$,
$\;E[MS_{\text{match}}] = \sigma^2_{\text{match}}$
<br><br>
with $c$ clubs per league and $n$ matches per club. Solving downward recovers each component
without an optimiser.
<br><br>
That is preferable here for a specific reason rather than on principle. An iterative mixed
model was tried first and produced convergence warnings on several features, which would have
left those components quietly unreliable. A closed-form estimate cannot fail to converge.
Where it can fail is by returning a negative variance when the true component sits at zero,
and those cases are reported rather than hidden, because a negative estimate is the
calculation saying the level explains nothing.
</div>

```python
def nested_components(df, value, outer='league', inner='team'):
    """Exact variance components for a balanced nested design."""
    d = df[[outer, inner, value]].dropna()
    grand = d[value].mean()
    inner_sizes = d.groupby([outer, inner]).size()
    n = len(inner_sizes) / (1.0 / inner_sizes).sum()          # harmonic mean, matches per club
    outer_counts = d.groupby(outer)[inner].nunique()
    c = len(outer_counts) / (1.0 / outer_counts).sum()        # harmonic mean, clubs per league

    inner_means = d.groupby([outer, inner])[value].mean()
    outer_means = d.groupby(outer)[value].mean()
    a, b_total, N = d[outer].nunique(), len(inner_sizes), len(d)

    ss_outer = sum(inner_sizes.loc[g].sum() * (outer_means[g] - grand) ** 2
                   for g in outer_means.index)
    ss_inner = sum(inner_sizes.loc[(g, t)] * (inner_means[(g, t)] - outer_means[g]) ** 2
                   for g, t in inner_means.index)
    fitted = d.set_index([outer, inner]).index.map(inner_means)
    ss_res = float(((d[value].to_numpy() - np.asarray(fitted)) ** 2).sum())

    ms_outer = ss_outer / (a - 1)
    ms_inner = ss_inner / (b_total - a)
    ms_res = ss_res / (N - b_total)

    s2_match = ms_res
    s2_club_raw = (ms_inner - ms_res) / n
    s2_league_raw = (ms_outer - ms_inner) / (n * c)
    neg = [k for k, v in (('club', s2_club_raw), ('league', s2_league_raw)) if v < 0]
    s2_club, s2_league = max(s2_club_raw, 0.0), max(s2_league_raw, 0.0)
    tot = s2_league + s2_club + s2_match

    return {'feature': value,
            'pct_league': 100 * s2_league / tot, 'pct_club': 100 * s2_club / tot,
            'pct_match': 100 * s2_match / tot,
            'sd_league': np.sqrt(s2_league), 'sd_club': np.sqrt(s2_club),
            'sd_match': np.sqrt(s2_match),
            'F_league': ms_outer / ms_inner, 'F_club': ms_inner / ms_res,
            'p_league': float(stats.f.sf(ms_outer / ms_inner, a - 1, b_total - a)),
            'p_club': float(stats.f.sf(ms_inner / ms_res, b_total - a, N - b_total)),
            'negative_estimate': ','.join(neg)}


VC_FEATURES = [f for f in ALL_FEATURES if f not in ('n_nodes', 'n_edges')]
varcomp = pd.DataFrame([nested_components(team_match, f) for f in VC_FEATURES])
varcomp['q_league'] = multipletests(varcomp.p_league, method='fdr_bh')[1]
varcomp['q_club'] = multipletests(varcomp.p_club, method='fdr_bh')[1]
varcomp['family'] = varcomp.feature.map(fam_of)

print(f"Mean variance share across {len(varcomp)} features")
print(f"   league {varcomp.pct_league.mean():5.1f}%")
print(f"   club   {varcomp.pct_club.mean():5.1f}%")
print(f"   match  {varcomp.pct_match.mean():5.1f}%")
print(f"\nSignificant at FDR 5%: league {(varcomp.q_league < 0.05).sum()} of {len(varcomp)}, "
      f"club {(varcomp.q_club < 0.05).sum()} of {len(varcomp)}")
print(f"Features returning a negative league component: "
      f"{(varcomp.negative_estimate.str.contains('league')).sum()}")
varcomp.sort_values('pct_club', ascending=False)[
    ['feature', 'family', 'pct_league', 'pct_club', 'pct_match', 'q_league', 'q_club']
].round(4)
```

Averaged over 20 features, league accounts for 2.2% of the variance in passing network

The split within that average is the informative part. Club matters most for volume and
tempo, with total passes at 49.8%, passes per possession at 48.6%, mean edge weight at 47.6%, long
sequence share at 45.4%. How much a side passes, and how long its possessions run, is close to

<br><br>
Where league does register, it is spatial rather than topological. The two largest league

<br><br>
Three quarters of the variance sitting at match level also carries a warning for the wider

</div>

<h3 style="color:#22414E;font-family:Helvetica,Arial,sans-serif;font-size:17px;
border-bottom:1.6px solid #CFD8DC;padding-bottom:7px;margin:26px 0 12px;">
<span style="color:#E8694A;">9.1</span>&nbsp;&nbsp;Uncertainty on the variance shares</h3>


The closed-form components of Section 9 are point estimates. A nested ANOVA gives no standard
error for a variance share, so the shares are resampled directly. Clubs are the resampling unit
rather than matches, because a club is the exchangeable object at the level where the club
component is estimated, and resampling matches would hold the set of clubs fixed and understate
the uncertainty in the very quantity of interest.


```python
def bootstrap_variance_shares(df, feature, n_boot=300, seed=SEED):
    """Percentile intervals for the three variance shares, resampling clubs."""
    rng = np.random.default_rng(seed)
    clubs_by_league = {lg: g.team.unique() for lg, g in df.groupby('league')}
    draws = []
    for _ in range(n_boot):
        parts = []
        for lg, clubs in clubs_by_league.items():
            for j, c in enumerate(rng.choice(clubs, len(clubs), replace=True)):
                sub = df[df.team == c].copy()
                sub['team'] = f'{c}__{j}'          # duplicated clubs act as distinct units
                parts.append(sub)
        boot = pd.concat(parts, ignore_index=True)
        try:
            r = nested_components(boot, feature)
            draws.append((r['pct_league'], r['pct_club'], r['pct_match']))
        except Exception:
            continue
    d = np.array(draws)
    return {'feature': feature,
            'league': d[:, 0].mean(), 'league lo': np.percentile(d[:, 0], 2.5),
            'league hi': np.percentile(d[:, 0], 97.5),
            'club': d[:, 1].mean(), 'club lo': np.percentile(d[:, 1], 2.5),
            'club hi': np.percentile(d[:, 1], 97.5),
            'match': d[:, 2].mean(), 'match lo': np.percentile(d[:, 2], 2.5),
            'match hi': np.percentile(d[:, 2], 97.5)}


BOOT_FEATURES = ['density', 'total_passes', 'strength_gini', 'final_third_share',
                 'weighted_reciprocity', 'progression_mean']
vc_boot = pd.DataFrame([bootstrap_variance_shares(team_match, f, n_boot=250)
                        for f in BOOT_FEATURES])
print("Variance shares with club-level bootstrap intervals, 250 resamples")
print(f"league share upper bound never exceeds "
      f"{vc_boot['league hi'].max():.1f}% across these features")
print(f"club share interval width averages "
      f"{(vc_boot['club hi'] - vc_boot['club lo']).mean():.1f} percentage points")
vc_boot.round(2)
```

<div style="background:#FDF2EC;border-left:4px solid #E8694A;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#E8694A;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Result</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">The league share is small with an interval that confirms it</h4>
Resampling clubs rather than matches gives the honest interval, and it is wider than the point
estimates alone suggest. Across six representative features the upper bound on the league share
reaches 19.3% at its highest, which is a reminder that four leagues is a small number of units
even when there are 3,034 matches beneath them. The club share is estimated far more precisely,
with intervals averaging 12.5 percentage points across features whose point estimates run from
roughly a fifth to a half.
<br><br>
This is a third independent route to the same conclusion, after the closed-form ANOVA and the
Bayesian posterior. The three differ in their assumptions. The ANOVA assumes balance and
normality, the Bayesian model assumes a prior, and the bootstrap assumes only that clubs are
exchangeable within a league. They agree, which is what makes the finding solid rather than an
artefact of one estimator.
</div>

<div id="sec10" style="border-left:5px solid #E8694A;padding:16px 0 16px 20px;
margin:30px 0 14px;font-family:Helvetica,Arial,sans-serif;">
<span style="color:#E8694A;font-size:12.5px;letter-spacing:2.6px;font-weight:bold;">SECTION 10</span>
<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Hierarchical Bayesian Decomposition</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">Putting intervals on the shares, and finding one that cannot be estimated</div></div>


The ANOVA gives point estimates. A hierarchical Bayesian version of the same model gives a
posterior for each variance, which matters here because one of the three components turns out
to be far less certain than the others, and a point estimate would conceal that.


```python
import pymc as pm
import arviz as az

BAYES_FEATURE = 'density'
d_b = team_match[['league', 'team', BAYES_FEATURE]].dropna()
lg = sorted(d_b.league.unique()); li = {g: i for i, g in enumerate(lg)}
cl = sorted(d_b.team.unique()); ci = {c: i for i, c in enumerate(cl)}
y_b = ((d_b[BAYES_FEATURE] - d_b[BAYES_FEATURE].mean()) / d_b[BAYES_FEATURE].std()).to_numpy()
li_idx = d_b.league.map(li).to_numpy(); ci_idx = d_b.team.map(ci).to_numpy()


def fit_variance_model(prior_scale, seed=SEED):
    """Non-centred three-level model. The centred form funnels when a scale is near zero."""
    with pm.Model():
        mu = pm.Normal('mu', 0, 1)
        sd_league = pm.HalfNormal('sd_league', prior_scale)
        sd_club = pm.HalfNormal('sd_club', 0.5)
        sd_match = pm.HalfNormal('sd_match', 1)
        z_lg = pm.Normal('z_lg', 0, 1, shape=len(lg))
        z_cl = pm.Normal('z_cl', 0, 1, shape=len(cl))
        pm.Normal('obs', mu + (z_lg * sd_league)[li_idx] + (z_cl * sd_club)[ci_idx],
                  sd_match, observed=y_b)
        return pm.sample(1500, tune=2000, chains=4, cores=1, target_accept=0.99,
                         random_seed=seed, progressbar=False)


sensitivity = []
for scale, label in [(0.25, 'HalfNormal(0.25)'), (0.5, 'HalfNormal(0.5)'),
                     (1.0, 'HalfNormal(1.0)')]:
    idata = fit_variance_model(scale)
    post = idata.posterior
    tot = post['sd_league']**2 + post['sd_club']**2 + post['sd_match']**2
    row = {'Prior on sd_league': label,
           'divergences': int(idata.sample_stats.diverging.sum())}
    for nm in ('league', 'club', 'match'):
        share = 100 * (post[f'sd_{nm}']**2) / tot
        row[f'{nm} %'] = float(share.mean())
        row[f'{nm} 94% HDI'] = (f"[{float(share.quantile(0.03)):.1f}, "
                                f"{float(share.quantile(0.97)):.1f}]")
    sensitivity.append(row)
prior_sensitivity = pd.DataFrame(sensitivity)
print(f"Three-level model on {BAYES_FEATURE}, {len(y_b)} observations, "
      f"{len(lg)} leagues, {len(cl)} clubs")
prior_sensitivity.round(2)
```

<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">League Signatures and What Survives</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">A second route to the same question, with different machinery</div></div>


The variance decomposition is an inferential answer. A predictive one asks something adjacent
and checks it independently. Given a single team-match network and nothing else, can a
classifier say which league it came from? If leagues have structural signatures, this should
be possible. If they do not, it should not be.

Folds are grouped by club, so no club appears on both sides of a split. Without that, a
classifier can memorise a specific team and infer its league from that, which measures club
recall rather than league structure and would inflate the result.


```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GroupKFold, cross_val_predict
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import balanced_accuracy_score, confusion_matrix

TOPO = FAMILIES['Topology'] + FAMILIES['Concentration'] + FAMILIES['Volume and tempo']
SPAT = FAMILIES['Spatial']
y_lg = team_match.league.to_numpy()
groups = team_match.team.to_numpy()

rows = []
preds = {}
for set_name, cols in [('topology and volume', TOPO), ('spatial only', SPAT),
                       ('all features', TOPO + SPAT)]:
    X = team_match[cols].fillna(team_match[cols].median()).to_numpy()
    for mdl_name, mdl in [
        ('logistic', make_pipeline(StandardScaler(), LogisticRegression(max_iter=3000))),
        ('random forest', RandomForestClassifier(n_estimators=400, random_state=SEED,
                                                 min_samples_leaf=3))]:
        pred = cross_val_predict(mdl, X, y_lg, cv=GroupKFold(n_splits=5), groups=groups)
        rows.append({'feature set': set_name, 'model': mdl_name,
                     'balanced accuracy': balanced_accuracy_score(y_lg, pred)})
        preds[(set_name, mdl_name)] = pred
league_clf = pd.DataFrame(rows)
chance = 1 / team_match.league.nunique()
print(f"Chance level with {team_match.league.nunique()} balanced classes: {chance:.3f}")
league_clf.round(4)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 5.2))

ax = axes[0]
lc = league_clf.sort_values('balanced accuracy')
y = np.arange(len(lc))
cols = [CORAL if 'all' in s else (TEAL if 'spatial' in s else SLATE)
        for s in lc['feature set']]
ax.barh(y, lc['balanced accuracy'], color=cols, edgecolor=SLATE, linewidth=0.6)
ax.axvline(chance, color=CORAL, ls='--', lw=1.8)
ax.text(chance + 0.004, 0.1, 'chance', fontsize=8.6, color=CORAL, rotation=90)
ax.set_yticks(y)
ax.set_yticklabels([f"{r['feature set']}\n{r['model']}" for _, r in lc.iterrows()],
                   fontsize=8)
ax.set_xlabel('Balanced accuracy, grouped 5-fold by club')
ax.set_xlim(0.2, max(0.45, lc['balanced accuracy'].max() * 1.12))
ax.set_title('(a) League is barely identifiable', fontsize=10.8)

ax = axes[1]
best = league_clf.loc[league_clf['balanced accuracy'].idxmax()]
cm = confusion_matrix(y_lg, preds[(best['feature set'], best['model'])],
                      labels=sorted(set(y_lg)), normalize='true')
im = ax.imshow(cm, cmap=CMAP_SEQ, vmin=0, vmax=max(0.6, cm.max()))
labels = [l.split()[0] for l in sorted(set(y_lg))]
ax.set_xticks(range(4)); ax.set_xticklabels(labels, fontsize=8.6, rotation=30)
ax.set_yticks(range(4)); ax.set_yticklabels(labels, fontsize=8.6)
for i in range(4):
    for j in range(4):
        ax.text(j, i, f'{cm[i, j]:.2f}', ha='center', va='center', fontsize=8.6,
                color='white' if cm[i, j] > 0.35 else SLATE)
ax.set_xlabel('predicted'); ax.set_ylabel('actual')
ax.set_title(f'(b) Best model, {best["feature set"]}', fontsize=10.8)
ax.grid(False)

ax = axes[2]
league_means = team_match.groupby('league')[
    ['final_third_share', 'origin_x_mean', 'progression_mean', 'density',
     'weighted_reciprocity', 'strength_gini']].mean()
z = (league_means - league_means.mean()) / league_means.std()
im2 = ax.imshow(z.T, cmap=CMAP_DIV, vmin=-1.8, vmax=1.8, aspect='auto')
ax.set_xticks(range(len(z)))
ax.set_xticklabels([l.split()[0] for l in z.index], fontsize=8.6, rotation=30)
ax.set_yticks(range(z.shape[1]))
ax.set_yticklabels([c.replace('_', ' ') for c in z.columns], fontsize=8.4)
ax.set_title('(c) Where leagues do differ, it is spatial', fontsize=10.8)
ax.grid(False)
plt.colorbar(im2, ax=ax, shrink=0.8, label='sd from the four-league mean')

fig.suptitle('Figure 3. A predictive test of the same question',
             fontsize=14, color=SLATE, fontweight='bold', y=1.02)
plt.tight_layout(); plt.show()
```

<span style="color:#E8694A;">11.1</span>&nbsp;&nbsp;Which leagues actually differ, and on what</h3>


Section 9 reports whether league matters at all. When it does, the natural follow-up is which
pairs of leagues differ, and that is a multiple-comparison problem with six pairs per feature.
Tukey's honest significant difference controls the family-wise error rate across all pairs
simultaneously, which a series of two-sample tests would not.

The unit of analysis is the club mean rather than the individual match, because clubs are nested
within leagues and treating 3,034 matches as independent observations of four leagues would
inflate the degrees of freedom enormously.


```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

club_level = team_match.groupby(['league', 'team'])[
    ['final_third_share', 'origin_x_mean', 'progression_mean', 'density',
     'weighted_reciprocity', 'strength_gini']].mean().reset_index()

tukey_rows = []
for f in ['final_third_share', 'origin_x_mean', 'progression_mean', 'density']:
    res = pairwise_tukeyhsd(club_level[f], club_level.league, alpha=0.05)
    frame = pd.DataFrame(res.summary().data[1:], columns=res.summary().data[0])
    for _, r in frame.iterrows():
        tukey_rows.append({'feature': f, 'pair': f"{r['group1']} vs {r['group2']}",
                           'difference': float(r['meandiff']),
                           'p adjusted': float(r['p-adj']),
                           'significant': bool(r['reject'])})
tukey = pd.DataFrame(tukey_rows)
sig = tukey[tukey.significant]
print(f"Tukey HSD on club means, {club_level.league.nunique()} leagues, "
      f"{len(club_level)} clubs")
print(f"pairwise comparisons: {len(tukey)} across 4 features")
print(f"significant at family-wise 5%: {len(sig)}")
if len(sig):
    for _, r in sig.iterrows():
        print(f"   {r['feature']:20s} {r['pair']:38s} "
              f"diff {r['difference']:+.4f}, p = {r['p adjusted']:.4f}")
tukey.round(4)
```

padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#E8694A;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Result</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">Only the spatial contrasts survive a family-wise correction</h4>
Across 24 pairwise comparisons spanning four features, nine are significant and every one of
them is spatial. Five separate leagues on final-third share and four on mean pass origin. Not a
single comparison on density or on forward progression survives the correction.
<br><br>
The largest contrast is Ligue 1 against the Premier League, which differ by 5.4 percentage points
in the share of passes played in the final third and by 3.6 metres in mean pass origin, both at
p below 0.0001. Ligue 1 sides play the deepest of the four leagues and Premier League sides the
highest, with La Liga and Serie A between them.
<br><br>
Using club means rather than individual matches is what makes this defensible. With 3,034
matches the same test would find almost everything significant, because it would be treating 38
matches by one club as 38 independent observations of its league. With 80 club means the degrees
of freedom reflect the actual number of independent units, and most of the apparent differences
disappear.
<br><br>
This is the third method to reach the conclusion of Section 11, after the variance decomposition
and the classifier, and the first to say precisely which leagues differ from which.

Section 9 found that the only features with a non-trivial league share describe location
rather than topology. This section looks at that directly, on a pitch, because a coefficient
of 12.7% for final-third share is hard to picture and a heat map is not.

Every location is in StatsBomb coordinates, a 120 by 80 pitch with the attacking direction
left to right, so all figures are oriented the same way regardless of which half a team
started in.


```python
from matplotlib.patches import Arc, Rectangle, Circle

PITCH_X, PITCH_Y = 120.0, 80.0

def draw_pitch(ax, line=None, fill=None, lw=1.3):
    """StatsBomb-coordinate pitch, drawn to scale."""
    line = line or '#9FB3BC'
    fill = PAPER if fill is None else fill
    if fill != 'none':
        ax.set_facecolor(fill)
    def R(x, y, w, h):
        ax.add_patch(Rectangle((x, y), w, h, fill=False, ec=line, lw=lw, zorder=1))
    R(0, 0, PITCH_X, PITCH_Y)
    ax.plot([60, 60], [0, 80], color=line, lw=lw, zorder=1)
    ax.add_patch(Circle((60, 40), 10, fill=False, ec=line, lw=lw, zorder=1))
    ax.add_patch(Circle((60, 40), 0.6, color=line, zorder=1))
    R(0, 18, 18, 44); R(102, 18, 18, 44)
    R(0, 30, 6, 20);  R(114, 30, 6, 20)
    ax.add_patch(Circle((12, 40), 0.6, color=line, zorder=1))
    ax.add_patch(Circle((108, 40), 0.6, color=line, zorder=1))
    ax.add_patch(Arc((12, 40), 20, 20, theta1=-53, theta2=53, color=line, lw=lw, zorder=1))
    ax.add_patch(Arc((108, 40), 20, 20, theta1=127, theta2=233, color=line, lw=lw, zorder=1))
    R(-2, 36, 2, 8); R(120, 36, 2, 8)
    ax.set_xlim(-4, 124); ax.set_ylim(-4, 84)
    ax.set_aspect('equal'); ax.axis('off')
    return ax


def parse_loc(v):
    if isinstance(v, str):
        try:
            return json.loads(v)
        except Exception:
            return None
    return v if isinstance(v, list) else None


def league_pass_locations(league, max_matches=90):
    """Pooled pass origins for one league, sampled over matches."""
    ids = (team_match[team_match.league == league]
           .drop_duplicates('match_id').match_id.head(max_matches))
    xs, ys, ex, ey = [], [], [], []
    for mid in ids:
        ev = pd.read_parquet(f"{DATA}/events/{int(mid)}.parquet",
                             columns=['type', 'pass_outcome', 'location', 'pass_end_location'])
        p = ev[(ev.type == 'Pass') & (ev.pass_outcome.isna())]
        for a, b in zip(p.location, p.pass_end_location):
            la, lb = parse_loc(a), parse_loc(b)
            if la and lb:
                xs.append(la[0]); ys.append(la[1]); ex.append(lb[0]); ey.append(lb[1])
    return np.array(xs), np.array(ys), np.array(ex), np.array(ey)


league_locs = {lg: league_pass_locations(lg) for lg in LEAGUE_COLOURS}
print("Pooled pass origins per league")
for lg, (x, y, _, _) in league_locs.items():
    print(f"  {lg:16s} {len(x):>7,} passes   mean x {x.mean():5.1f}   "
          f"final third {100*(x >= 80).mean():4.1f}%")
```

```python
fig, axes = plt.subplots(2, 4, figsize=(18, 9))

for k, (lg, (x, y, ex, ey)) in enumerate(league_locs.items()):
    # top row: where passes start
    ax = axes[0, k]
    ax.set_facecolor(PAPER)
    # a smoothed density reads better than raw hexbin counts, which are dominated by
    # the handful of cells around the centre circle
    from scipy.stats import gaussian_kde
    sub = np.random.default_rng(SEED).choice(len(x), min(9000, len(x)), replace=False)
    kde = gaussian_kde(np.vstack([x[sub], y[sub]]), bw_method=0.28)
    gx, gy = np.mgrid[0:120:120j, 0:80:80j]
    dens = kde(np.vstack([gx.ravel(), gy.ravel()])).reshape(gx.shape)
    ax.imshow(dens.T, origin='lower', extent=(0, 120, 0, 80), cmap=CMAP_SEQ,
              aspect='auto', zorder=1)
    draw_pitch(ax, line='#6E8A95', fill='none')      # lines drawn over the density
    ax.set_title(f'{lg}\nmean origin x = {x.mean():.1f}', fontsize=10.6, color=SLATE)

    # bottom row: net flow, where passes end minus where they start
    ax = axes[1, k]
    ax.set_facecolor(PAPER)
    H_s, xe, ye = np.histogram2d(x, y, bins=[12, 8], range=[[0, 120], [0, 80]])
    H_e, _, _ = np.histogram2d(ex, ey, bins=[12, 8], range=[[0, 120], [0, 80]])
    net = (H_e - H_s) / max(H_s.sum(), 1) * 100
    lim = np.abs(net).max()
    ax.imshow(net.T, origin='lower', extent=(0, 120, 0, 80), cmap=CMAP_DIV,
              vmin=-lim, vmax=lim, zorder=1, aspect='auto')
    draw_pitch(ax, line=SLATE, fill='none', lw=1.0)
    ax.set_title('net flow into each zone\nteal loses passes, orange gains them',
                 fontsize=9.6, color=SLATE)

fig.suptitle('Figure 4. Where each league plays, and which way the ball moves\n'
             'top row is pass origin density, bottom row is destinations minus origins',
             fontsize=13.5, color=SLATE, fontweight='bold', y=1.0)
plt.tight_layout(); plt.show()
```

<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Player Roles and Network Position</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">Who holds a passing network together, and whether that role is a club property</div></div>


A team-level metric averages over players and hides the structure it is summarising. Two sides
with identical centralization can arrive there differently, one through a dominant deep
playmaker and the other through a pair of centre backs who circulate between themselves.

The player table carries 41,807 rows for the cross section, one per player per team-match,
with degree, strength, betweenness, eigenvector centrality, PageRank and involvement share.


```python
players['involvement_rank'] = (players.groupby(['match_id', 'team'])
                               .involvement_share.rank(ascending=False))
pl = players.merge(team_match[['match_id', 'team', 'league']],
                   on=['match_id', 'team'], how='left')

# how concentrated is passing involvement across the ranked players
rank_profile = (pl[pl.involvement_rank <= 14]
                .groupby(['league', 'involvement_rank']).involvement_share.mean().unstack(0))

# does a player hold the same network position from match to match?
counts = pl.groupby('player').size()
regulars = counts[counts >= 15].index
reg = pl[pl.player.isin(regulars)]
consistency = []
for metric in ['involvement_share', 'eigenvector', 'betweenness', 'pagerank']:
    d = reg[['player', metric]].dropna()
    pm = d.groupby('player')[metric].transform('mean')
    consistency.append({'metric': metric,
                        'between-player R2': float(np.corrcoef(d[metric], pm)[0, 1] ** 2),
                        'players': int(d.player.nunique())})
consistency = pd.DataFrame(consistency)
print(f"Players with at least 15 team-matches: {len(regulars):,}")
print("How much of a player's network position repeats across matches")
print(consistency.round(3).to_string(index=False))
print(f"\nTop involvement share in a single team-match: "
      f"{pl.involvement_share.max():.3f}")
print(f"Median share held by the most involved player: "
      f"{pl[pl.involvement_rank == 1].involvement_share.median():.3f}")
```

<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Network Sensitivity Analysis</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">Every result so far rests on four construction choices, so vary them</div></div>


Section 6 made four decisions about what counts as an edge. Each was defensible and each was
arbitrary at the margin. A conclusion that depends on those choices is a conclusion about the
choices.

Three alternatives are computed for every team-match network alongside the default, namely set pieces
included, the graph treated as undirected, and a minimum edge weight of three passes. The
question is whether the headline result and the individual metrics survive.


```python
ALT_SPECS = {'sp': 'set pieces included', 'undir': 'undirected',
             'thr3': 'edge weight at least 3'}
SENS_FEATURES = ['density', 'pass_entropy_norm', 'strength_gini',
                 'betweenness_centralization']

rows = []
for f in SENS_FEATURES:
    for key, label in ALT_SPECS.items():
        alt = f'{f}__{key}'
        if alt not in team_match.columns:
            continue
        vc_def = nested_components(team_match, f)
        vc_alt = nested_components(team_match, alt)
        rows.append({
            'feature': f, 'specification': label,
            'default mean': team_match[f].mean(),
            'alternative mean': team_match[alt].mean(),
            'rank correlation': stats.spearmanr(team_match[f], team_match[alt])[0],
            'league % default': vc_def['pct_league'],
            'league % alternative': vc_alt['pct_league'],
            'club % default': vc_def['pct_club'],
            'club % alternative': vc_alt['pct_club'],
        })
sensitivity = pd.DataFrame(rows)
fragile = sensitivity[sensitivity['rank correlation'] < 0.7]
print(f"{len(sensitivity)} feature-specification combinations tested")
print(f"rank correlation with the default below 0.7: {len(fragile)}")
if len(fragile):
    # column names contain spaces, so itertuples renames them positionally; use the
    # frame directly rather than an attribute whose position is easy to miscount
    for _, r in fragile.iterrows():
        print(f"   {r['feature']} under '{r['specification']}': "
              f"rho = {r['rank correlation']:.3f}")
print(f"\nleague share stays below 5% in "
      f"{(sensitivity['league % alternative'] < 5).sum()} of {len(sensitivity)} alternatives")
sensitivity.round(4)
```

<div style="background:#FDF2EC;border-left:4px solid #E8694A;border-radius:3px;

color:#2F4750;font-size:14px;"><div style="color:#E8694A;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Result</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">The headline survives; one metric does not</h4>
The league share stays near zero under every alternative construction. Whatever choices are
made about set pieces, direction or thresholding, national leagues account for a negligible
part of the variation in passing structure. That is the result the project turns on and it is
not an artefact of how the graph was defined.
<br><br>
Betweenness centralization is a different matter. Its rank correlation with the default is
0.88 when set pieces are added, falls to 0.66 when the graph is undirected, and reaches
<strong>minus 0.04</strong> under a minimum edge weight of three. A near-zero correlation means
the thresholded version is measuring something unrelated to the unthresholded one. Removing
low-frequency pairings removes exactly the bridging edges that betweenness depends on, and
what survives is a different quantity wearing the same name.
<br><br>
Density under thresholding shifts the variance structure as well, with the club share rising
from 23.6% to 40.9%. Once rare pairings are stripped out, what remains is closer to a stable
club property, which is interpretable but is no longer the same measurement.
<br><br>
The practical consequence is that betweenness centralization is reported in this notebook as
descriptive and is not used to support any conclusion. Anyone comparing centralization figures
across published football network studies should check the threshold first, because the metric
is not comparable across them.
</div>

<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Networks and Match Outcomes</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">Whether structure relates to results, and what happens when quality is controlled</div></div>


This is the question most applied passing-network work is ultimately about. Teams that circulate
the ball well are assumed to win more, and the correlations in raw data support that. The
difficulty is that good teams both pass well and win, so an association between the two may be
measuring neither direction of influence.

Five specifications are fitted for each feature, each removing a different confounder, so that
what survives can be attributed to something other than team quality.


<div style="background:#FCFAF5;border:1px solid #CFD8DC;border-radius:3px;
padding:16px 20px;margin:14px 0;text-align:center;">$$\text{GD}_{it} = \alpha + \beta z_{it} + \gamma\,\text{home}_{it}
+ \delta_1 q_i + \delta_2 q_{o(it)} + \varepsilon_{it}$$</div>

<div style="background:#EEF3F5;border-left:4px solid #22414E;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#22414E;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Method</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">The five specifications and what each one removes</h4>
<strong>Raw.</strong> Goal difference on the standardised network feature and home advantage
only. This is the association most analyses report.
<br><br>
<strong>Quality controlled.</strong> Adds the team's own season strength and the opponent's. Own
strength is computed leave-one-out, so the match being modelled never contributes to the control
that predicts it, which would otherwise build the outcome into the covariate.
<br><br>
<strong>Club fixed effects.</strong> Compares a club against itself across its own 38 matches,
removing everything stable about that club including squad quality and managerial approach.
<br><br>
<strong>Mixed effects.</strong> Replaces the fixed effects with club random intercepts, which
pools information across clubs instead of discarding between-club variation entirely, and
reports how much residual variance sits at club level.
<br><br>
<strong>Poisson.</strong> Goals scored are counts, not a continuous quantity, and modelling them
with OLS assumes constant variance that count data do not have. A Poisson model with a log link
is the correct family, and overdispersion is checked before it is trusted.
<br><br>
Standard errors are clustered on the match throughout, because the two teams in a match produce
mechanically linked outcomes and treating 3,034 rows as independent would understate every
interval.
</div>

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

om = team_match.copy()
om['gd'] = om.goals - om.goals_against
om['pts'] = np.where(om.gd > 0, 3, np.where(om.gd == 0, 1, 0))
tot_pts = om.groupby('team').pts.transform('sum')
n_games = om.groupby('team').pts.transform('size')
om['quality'] = (tot_pts - om.pts) / (n_games - 1)          # leave one out
om['opp_quality'] = om.opponent.map(om.groupby('team').pts.mean())

OUTCOME_FEATURES = ['density', 'weighted_reciprocity', 'strength_gini',
                    'pass_entropy_norm', 'betweenness_centralization', 'clustering',
                    'total_passes', 'passes_per_possession', 'progression_mean',
                    'final_third_share', 'zone_entropy_norm', 'pass_length_mean']

rows = []
for f in OUTCOME_FEATURES:
    d = om.assign(z=(om[f] - om[f].mean()) / om[f].std())
    cl = {'groups': d.match_id}
    m_raw = smf.ols("gd ~ z + is_home", d).fit(cov_type='cluster', cov_kwds=cl)
    m_qual = smf.ols("gd ~ z + is_home + quality + opp_quality", d).fit(
        cov_type='cluster', cov_kwds=cl)
    m_fe = smf.ols("gd ~ z + is_home + quality + opp_quality + C(team)", d).fit(
        cov_type='cluster', cov_kwds=cl)
    with warnings.catch_warnings():
        warnings.simplefilter('ignore')
        m_me = smf.mixedlm("gd ~ z + is_home + quality + opp_quality", d,
                           groups=d['team']).fit(reml=False)
    m_pois = smf.glm("goals ~ z + is_home + quality + opp_quality", d,
                     family=sm.families.Poisson()).fit(cov_type='cluster', cov_kwds=cl)
    rows.append({
        'feature': f,
        'raw': m_raw.params['z'],
        'quality controlled': m_qual.params['z'],
        'club fixed effects': m_fe.params['z'],
        'mixed effects': m_me.params['z'],
        'Poisson (log goals)': m_pois.params['z'],
        'se (quality)': m_qual.bse['z'],
        'p (quality)': m_qual.pvalues['z'],
        'sign flip': np.sign(m_raw.params['z']) != np.sign(m_qual.params['z']),
        'shrinkage %': 100 * (1 - abs(m_qual.params['z']) /
                              max(abs(m_raw.params['z']), 1e-9)),
    })
outcomes = pd.DataFrame(rows)
outcomes['q'] = multipletests(outcomes['p (quality)'], method='fdr_bh')[1]
outcomes = outcomes.sort_values('quality controlled', key=abs, ascending=False)

print(f"{len(outcomes)} features, five specifications each, "
      f"match-clustered standard errors")
print(f"sign reversals once quality is controlled: "
      f"{int(outcomes['sign flip'].sum())} of {len(outcomes)}")
print(f"median shrinkage from controlling quality: "
      f"{outcomes['shrinkage %'].median():.0f}%")
print(f"surviving FDR 5% after quality control: {int((outcomes.q < 0.05).sum())}")
outcomes.round(4)
```

```python
# Is Poisson defensible for goals, or is the count overdispersed?
d0 = om.assign(z=(om.density - om.density.mean()) / om.density.std())
pois = smf.glm("goals ~ z + is_home + quality + opp_quality", d0,
               family=sm.families.Poisson()).fit()
pearson_chi2 = float((pois.resid_pearson ** 2).sum())
dispersion = pearson_chi2 / pois.df_resid
nb = smf.glm("goals ~ z + is_home + quality + opp_quality", d0,
             family=sm.families.NegativeBinomial(alpha=1.0)).fit()
print("Choosing a count family for goals scored")
print(f"  mean goals {om.goals.mean():.3f}, variance {om.goals.var():.3f}")
print(f"  Pearson dispersion statistic {dispersion:.3f}")
print(f"  {'no evidence of overdispersion, Poisson is appropriate' if dispersion < 1.25 else 'overdispersed, negative binomial preferred'}")
print(f"  Poisson AIC {pois.aic:.1f} | negative binomial AIC {nb.aic:.1f}")

# Does the association differ across the outcome distribution?
from statsmodels.regression.quantile_regression import QuantReg
X = sm.add_constant(d0[['z', 'is_home', 'quality', 'opp_quality']].astype(float))
qr_rows = []
for tau in [0.15, 0.30, 0.50, 0.70, 0.85]:
    qr = QuantReg(d0.gd.astype(float), X).fit(q=tau)
    qr_rows.append({'quantile': tau, 'density coefficient': qr.params['z'],
                    'se': qr.bse['z']})
quantiles = pd.DataFrame(qr_rows)
print("\nQuantile regression, density against goal difference")
print("asks whether the relationship is the same for heavy defeats and heavy wins")
quantiles.round(4)
```

Total passes correlates with goal difference at +0.331 in the raw specification, which is the

<br><br>
Panel (c) shows why. Every feature correlates more strongly with team quality than with the

<br><br>
With club fixed effects, comparing a side against its own 38 matches, the coefficients collapse

matters because fixed effects discard between-club information entirely while random intercepts
pool it, and the two approaches answering the same way is stronger than either alone.

The quantile regression adds something the mean models cannot. The density coefficient is not
constant across the outcome distribution, so the relationship between structure and result
differs between heavy defeats and comfortable wins. That is consistent with game state driving
passing behaviour rather than the reverse, since a team protecting a lead plays differently from

appropriate family, and the dispersion statistic confirms it is defensible rather than assumed.
<br><br>
The negative binomial alternative is fitted alongside it for comparison. Where the dispersion
statistic sits near one, the extra parameter buys nothing and the simpler model is preferred,
which is the reason the check is run before the model is trusted rather than after.
<br><br>

scorelines being close to Poisson is a long-standing result in the sport's statistics, and
recovering it here is a further sign the reconstructed goal counts are correct.

Everything so far treats a passing network as a summary of a completed match. That is what the
statistics require, and it discards the thing football is actually made of, which is sequence.
A network with 130 edges was assembled one pass at a time over ninety minutes, and the order
matters to how the structure came about.

Three animations follow, each answering something the static figures cannot. The first replays
a single possession as it happened. The second watches a full match network assemble from its
first pass to its last. The third holds a club fixed and morphs between the average networks of
different periods, which is what a structural change looks like when you can see it move.

All three are produced from the same event data as every number above, and are embedded as
animated GIF so they display in any notebook viewer. MP4 copies are written alongside them for
anyone who wants to scrub through frame by frame.


```python
import matplotlib.animation as animation
import base64
from IPython.display import HTML, display

# Animations are written as animated GIF rather than MP4.
#
# An HTML5 video element carrying a base64 data URI renders locally and is blocked by
# the Kaggle notebook viewer, which shows an unplayable black rectangle instead. An
# animated GIF arrives through an img tag, which the viewer does render, and at these
# frame counts it is also the smaller file. MP4 copies are written alongside where
# ffmpeg is available, for anyone who wants to scrub through them.
try:
    import imageio_ffmpeg
    plt.rcParams['animation.ffmpeg_path'] = imageio_ffmpeg.get_ffmpeg_exe()
    MP4_OK = True
except Exception:
    MP4_OK = False


def save_animation(anim, path, fps=5, dpi=85):
    """Write the GIF that will be displayed, and an MP4 copy if ffmpeg is present."""
    gif = path.rsplit('.', 1)[0] + '.gif'
    anim.save(gif, writer=animation.PillowWriter(fps=fps), dpi=dpi)
    if MP4_OK:
        try:
            anim.save(path, writer=animation.FFMpegWriter(
                fps=fps, bitrate=2200, extra_args=['-pix_fmt', 'yuv420p']), dpi=dpi)
        except Exception:
            pass
    return gif


def show_video(path, width=780, caption=None):
    """Inline the animation so the notebook carries its own media."""
    data = base64.b64encode(open(path, 'rb').read()).decode()
    if caption:
        display(HTML(f"<div style='font-family:Helvetica,Arial,sans-serif;font-size:13px;"
                     f"color:{SLATE};margin:8px 0 4px;'><strong>{caption}</strong></div>"))
    display(HTML(f'<img src="data:image/gif;base64,{data}" width="{width}" '
                 f'style="border-radius:4px;">'))
    mp4 = path.rsplit('.', 1)[0] + '.mp4'
    extra = (f", MP4 copy {os.path.getsize(mp4)/1024:.0f} kB"
             if os.path.exists(mp4) else "")
    print(f"   {os.path.basename(path)}, {os.path.getsize(path)/1024:.0f} kB{extra}")


VIDEO_DIR = '/kaggle/working/videos'
os.makedirs(VIDEO_DIR, exist_ok=True)
print(f"MP4 copies alongside the GIFs: {MP4_OK}")
```

```python
def animate_possession(events, team, possession_id, fps=3, title=None):
    """Replay one possession pass by pass, in the order the passes were played."""
    p = events[(events.possession == possession_id) & (events.team == team)
               & (events.type == 'Pass') & (events.pass_outcome.isna())].sort_values('index')
    starts = [parse_loc(v) for v in p.location]
    ends = [parse_loc(v) for v in p.pass_end_location]
    keep = [i for i in range(len(p)) if starts[i] and ends[i]]
    if len(keep) < 4:
        return None
    starts = [starts[i] for i in keep]; ends = [ends[i] for i in keep]
    names = [str(p.player.iloc[i]).split()[-1] for i in keep]

    fig, ax = plt.subplots(figsize=(9.6, 6.4))
    draw_pitch(ax)
    if title:
        ax.set_title(title, fontsize=12, color=SLATE)

    def frame(k):
        if k == 0:
            return []
        i = k - 1
        x1, y1 = starts[i]; x2, y2 = ends[i]
        shade = CMAP_SEQ(0.35 + 0.6 * i / max(len(starts) - 1, 1))
        ax.annotate('', xy=(x2, y2), xytext=(x1, y1), zorder=3,
                    arrowprops=dict(arrowstyle='-|>', lw=2.4, color=shade, alpha=0.95))
        ax.scatter([x1], [y1], s=165, color=SLATE, edgecolor='white',
                   linewidth=1.5, zorder=4)
        ax.text(x1, y1 - 3.4, names[i], ha='center', va='top', fontsize=8,
                color=SLATE, zorder=5)
        if i == len(starts) - 1:
            ax.scatter([x2], [y2], s=260, marker='*', color=CORAL,
                       edgecolor='white', linewidth=1.3, zorder=6)
        return []

    return animation.FuncAnimation(fig, frame, frames=len(starts) + 1,
                                   interval=1000 / fps, blit=False, repeat=False)


# pick the longest uninterrupted passing move in the corpus, objectively
best = None
for mid in team_match.drop_duplicates('match_id').match_id.head(160):
    ev_ = pd.read_parquet(f"{DATA}/events/{int(mid)}.parquet")
    cnt = (ev_[(ev_.type == 'Pass') & (ev_.pass_outcome.isna())]
           .groupby(['possession', 'team']).size())
    if len(cnt) and (best is None or cnt.max() > best[0]):
        pid, tm_ = cnt.idxmax()
        best = (int(cnt.max()), int(mid), int(pid), tm_)

n_pass, mid, pid, team_name = best
ev = pd.read_parquet(f"{DATA}/events/{mid}.parquet")
meta_row = matches_meta[matches_meta.match_id == mid].iloc[0]
print(f"Longest single possession found: {n_pass} consecutive completed passes")
print(f"   {team_name}, {meta_row.home_team} against {meta_row.away_team}, "
      f"{str(meta_row.match_date)[:10]}")

anim1 = animate_possession(ev, team_name, pid, fps=3,
                           title=f'{team_name}: {n_pass} consecutive passes')
if anim1 is not None:
    path1 = save_animation(anim1, f'{VIDEO_DIR}/possession.mp4', fps=3)
    plt.close('all')
    show_video(path1, caption='One possession, replayed in the order it was played')
```

The possession above is the longest uninterrupted passing move in the sampled matches. Watched

return when the probe finds nothing.
<br><br>
Every network statistic in this notebook reduces that to a set of edge weights, and in doing so
discards the order entirely. Two teams can produce identical adjacency matrices from completely
different sequences, one building patiently and the other recycling possession after failed
attacks. The metrics cannot separate them, which is a real limit on what the analysis can
claim about style rather than about structure.

```python
def animate_network_growth(passes, positions, n_frames=36, fps=6, title=None):
    """Watch the passing graph assemble over the course of a match."""
    p = passes.sort_values('index').reset_index(drop=True)
    pos = {r.player: (r.x, r.y) for r in positions.itertuples()}
    cuts = np.linspace(8, len(p), n_frames).astype(int)
    fig, ax = plt.subplots(figsize=(9.6, 6.4))

    def frame(k):
        ax.clear()
        draw_pitch(ax)
        sub = p.iloc[:cuts[k]]
        e = sub.groupby(['player', 'pass_recipient']).size().reset_index(name='w')
        wmax = max(e.w.max(), 1)
        for r in e.itertuples():
            if r.player in pos and r.pass_recipient in pos:
                x1, y1 = pos[r.player]; x2, y2 = pos[r.pass_recipient]
                ax.annotate('', xy=(x2, y2), xytext=(x1, y1), zorder=2,
                            arrowprops=dict(arrowstyle='-', color=SLATE,
                                            lw=0.5 + 3.2 * r.w / wmax,
                                            alpha=0.15 + 0.55 * r.w / wmax,
                                            shrinkA=8, shrinkB=8))
        inv = sub.groupby('player').size()
        sz = np.array([110 + 950 * inv.get(pl_, 0) / max(inv.max(), 1)
                       for pl_ in positions.player])
        ax.scatter(positions.x, positions.y, s=sz, c=CORAL, edgecolor='white',
                   linewidth=1.6, zorder=3)
        for r in positions.itertuples():
            ax.text(r.x, r.y - 4.4, str(r.player).split()[-1], ha='center', va='top',
                    fontsize=7, color=SLATE, zorder=4)
        dens = len(e) / max(len(pos) * (len(pos) - 1), 1)
        ax.set_title(f"{title}    {cuts[k]} of {len(p)} passes,  density {dens:.2f}",
                     fontsize=11.5, color=SLATE)
        return []

    return animation.FuncAnimation(fig, frame, frames=len(cuts),
                                   interval=1000 / fps, blit=False, repeat=True)


def team_positions(passes):
    locs = passes.location.apply(parse_loc)
    ok = locs.notna()
    return (pd.DataFrame({'player': passes.player[ok],
                          'x': [l[0] for l in locs[ok]],
                          'y': [l[1] for l in locs[ok]]})
            .groupby('player').agg(x=('x', 'mean'), y=('y', 'mean')).reset_index())


# the densest network in the corpus, so the assembly is worth watching
row_dense = team_match.loc[team_match.density.idxmax()]
ev_d = pd.read_parquet(f"{DATA}/events/{int(row_dense.match_id)}.parquet")
p_d = passes_for_team(ev_d, row_dense.team)
pos_d = team_positions(p_d)
print(f"Densest network in the corpus: {row_dense.team} against {row_dense.opponent}, "
      f"density {row_dense.density:.3f}, {int(row_dense.total_passes)} passes")

anim2 = animate_network_growth(p_d, pos_d, n_frames=36, fps=6, title=row_dense.team)
path2 = save_animation(anim2, f'{VIDEO_DIR}/network_growth.mp4', fps=6)
plt.close('all')
show_video(path2, caption='A passing network assembling itself over ninety minutes')
```

```python
TOP_N_PLAYERS = 13     # eleven starters plus the two most used substitutes


def average_network(frame_rows, max_matches=14):
    """Average positions and edge weights across a set of team-matches."""
    pos_acc, edge_acc = [], []
    for r in frame_rows.head(max_matches).itertuples():
        ev_ = pd.read_parquet(f"{DATA}/events/{int(r.match_id)}.parquet")
        p_ = passes_for_team(ev_, r.team)
        if len(p_) < 50:
            continue
        pos_acc.append(team_positions(p_))
        edge_acc.append(p_.groupby(['player', 'pass_recipient']).size()
                        .reset_index(name='w'))
    if not pos_acc:
        return None, None
    pos = (pd.concat(pos_acc).groupby('player')
           .agg(x=('x', 'mean'), y=('y', 'mean'), n=('x', 'size')).reset_index())
    edges = (pd.concat(edge_acc).groupby(['player', 'pass_recipient']).w.mean()
             .reset_index())
    # Averaging over a season pools every rotation player, which produces a graph of
    # twenty-plus nodes that no team ever fielded. Restrict to the core group by total
    # involvement, which is what an average passing network conventionally shows.
    involvement = (edges.groupby('player').w.sum()
                   .add(edges.groupby('pass_recipient').w.sum(), fill_value=0))
    core = involvement.nlargest(TOP_N_PLAYERS).index
    pos = pos[pos.player.isin(core) & (pos.n >= max(2, len(pos_acc) // 3))]
    edges = edges[edges.player.isin(pos.player) & edges.pass_recipient.isin(pos.player)]
    return pos, edges


def animate_transition(states, fps=1.4, hold=4, title=None):
    """Morph between average networks, one frame block per state."""
    seq = [s for s in states for _ in range(hold)]
    fig, ax = plt.subplots(figsize=(9.6, 6.6))

    def frame(k):
        ax.clear()
        label, pos, edges = seq[k]
        draw_pitch(ax)
        pmap = {r.player: (r.x, r.y) for r in pos.itertuples()}
        e = edges.nlargest(30, 'w'); wmax = max(e.w.max(), 1)
        for r in e.itertuples():
            if r.player in pmap and r.pass_recipient in pmap:
                x1, y1 = pmap[r.player]; x2, y2 = pmap[r.pass_recipient]
                ax.annotate('', xy=(x2, y2), xytext=(x1, y1), zorder=2,
                            arrowprops=dict(arrowstyle='-|>', color=SLATE,
                                            lw=0.6 + 3.4 * r.w / wmax,
                                            alpha=0.22 + 0.55 * r.w / wmax,
                                            shrinkA=9, shrinkB=9,
                                            connectionstyle='arc3,rad=0.08'))
        strength = (edges.groupby('player').w.sum()
                    .add(edges.groupby('pass_recipient').w.sum(), fill_value=0))
        sz = np.array([150 + 1150 * strength.get(pl_, 0) / max(strength.max(), 1)
                       for pl_ in pos.player])
        ax.scatter(pos.x, pos.y, s=sz, c=CORAL, edgecolor='white', linewidth=1.7,
                   zorder=3)
        for r, s_ in zip(pos.itertuples(), sz):
            ax.text(r.x, r.y - 3.4 - s_ / 900, str(r.player).split()[-1],
                    ha='center', va='top', fontsize=7.4, color=SLATE, zorder=4)
        ax.set_title(f"{title}    {label}", fontsize=12, color=SLATE)
        return []

    return animation.FuncAnimation(fig, frame, frames=len(seq),
                                   interval=1000 / fps, blit=False, repeat=True)


# the spectator experiment, seen rather than tabulated: one club, crowd then no crowd
panel_club = (panel.groupby('team').league.nunique().sort_values(ascending=False)
              .index[0])
states = []
for season, label in [('FAWSL 2018/19', 'with a crowd, 2018/19'),
                      ('FAWSL 2020/21', 'behind closed doors, 2020/21'),
                      ('FAWSL 2023/24', 'crowd restored, 2023/24')]:
    rows_ = panel[(panel.team == panel_club) & (panel.league == season)]
    if len(rows_) < 4:
        continue
    pos_s, edges_s = average_network(rows_)
    if pos_s is not None:
        states.append((label, pos_s, edges_s))

print(f"Club with the fullest presence across seasons: {panel_club}")
for label, pos_s, _ in states:
    print(f"   {label}: {len(pos_s)} players in the average network")

if len(states) >= 2:
    anim3 = animate_transition(states, fps=1.4, hold=4, title=panel_club)
    path3 = save_animation(anim3, f'{VIDEO_DIR}/spectator_transition.mp4', fps=1.4)
    plt.close('all')
    show_video(path3, caption='The same club across the spectator transition')
```

The third animation shows one club's average passing network in three periods, and the shapes

Section 17 fits a trend through the crowd seasons and finds the closed-door season sits on it

margin:30px 0 14px;font-family:Helvetica,Arial,sans-serif;">
<span style="color:#E8694A;font-size:12.5px;letter-spacing:2.6px;font-weight:bold;">SECTION 17</span>
<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">The Spectator Absence Case Study</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">A treated season bracketed on both sides, and why that is still not enough</div></div>


The FA Women's Super League played its entire 2020/21 season behind closed doors, with two
seasons of crowd football before it and one after. That bracketing is what makes it worth
examining, because an interruption followed by a return to baseline is stronger evidence than an
interruption alone.

The identification problem is stated before the estimates rather than after them, because it
determines how the numbers should be read.


<div style="background:#FBEFE8;border-left:4px solid #C1613C;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#C1613C;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Caution</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">Three limits, one of which is fatal to hypothesis testing</h4>
<strong>Treatment is assigned at season level and there are four seasons.</strong> Whatever the
sample of matches, the number of independent units receiving or not receiving the treatment is
four. A permutation test that reassigns which season was treated therefore has a null
distribution with four values, and the observed estimate is one of them, so the smallest
attainable p-value is 0.4. No effect of any size can be rejected. Section 19 demonstrates this
by injecting shifts of up to two standard deviations and failing to detect any of them.
<br><br>
<strong>The season before treatment is incomplete.</strong> 2019/20 was abandoned in February
2020 after 87 of 132 matches, so it is the first half of a season rather than a season, and
first and second halves differ systematically.
<br><br>
<strong>The archive omits the two seasons after treatment.</strong> 2021/22 and 2022/23 are not
released, so the return to crowd football is observed three years later rather than immediately.
<br><br>
This section is therefore a bounded case study reporting descriptive contrasts with honest
intervals, not a causal estimate.
</div>

```python
panel['season'] = panel.league.str.replace('FAWSL ', '', regex=False)
SEASON_POS = {'2018/19': 0, '2019/20': 1, '2020/21': 2, '2023/24': 5}
panel['t'] = panel.season.map(SEASON_POS)
panel['closed'] = (panel.season == '2020/21').astype(int)

setup = (panel.groupby('season')
         .agg(networks=('match_id', 'size'), matches=('match_id', 'nunique'),
              clubs=('team', 'nunique'))
         .assign(crowd=lambda d: ['yes', 'yes', 'no', 'yes']))
print("FA Women's Super League, the four released seasons")
print(setup.to_string())
print(f"\ntreated networks {int(panel.closed.sum())}, control {int((1 - panel.closed).sum())}")

SPEC_FEATURES = ['density', 'weighted_reciprocity', 'strength_gini', 'pass_entropy_norm',
                 'betweenness_centralization', 'clustering', 'total_passes',
                 'passes_per_possession', 'progression_mean', 'final_third_share',
                 'zone_entropy_norm', 'pass_length_mean', 'long_sequence_share',
                 'avg_path_length']

season_means = panel.groupby('season')[SPEC_FEATURES].mean().T
season_means['direction'] = np.where(
    season_means['2023/24'] > season_means['2018/19'], 'rising', 'falling')
print("\nSeason means, with the treated season in the middle of the sequence")
season_means.round(3)
```

```python
def trend_adjusted_effect(frame, feature, treated_t=2):
    """Deviation of the treated season from a line fitted through the control seasons.

    A simple before-and-after difference attributes any secular change to the treatment.
    Fitting a trend through the seasons that were not treated, then measuring how far the
    treated season falls from it, separates the two.
    """
    ctrl = frame[frame.t != treated_t]
    cm = ctrl.groupby('t')[feature].mean()
    if len(cm) < 2:
        return np.nan
    slope, intercept = np.polyfit(cm.index.values.astype(float), cm.values, 1)
    return frame.loc[frame.t == treated_t, feature].mean() - (intercept + slope * treated_t)


rng_boot = np.random.default_rng(SEED)
idx_by_season = {t: panel.index[panel.t == t].to_numpy() for t in sorted(panel.t.unique())}

rows = []
for f in SPEC_FEATURES:
    naive = (panel.loc[panel.closed == 1, f].mean()
             - panel.loc[panel.closed == 0, f].mean())
    adj = trend_adjusted_effect(panel, f)
    sd = panel[f].std()
    draws = []
    for _ in range(600):
        take = np.concatenate([rng_boot.choice(v, len(v), replace=True)
                               for v in idx_by_season.values()])
        draws.append(trend_adjusted_effect(panel.loc[take], f))
    lo, hi = np.percentile(draws, [2.5, 97.5])
    rows.append({'feature': f, 'naive difference': naive, 'trend adjusted': adj,
                 'effect (% of sd)': 100 * adj / sd,
                 'CI low (% of sd)': 100 * lo / sd, 'CI high (% of sd)': 100 * hi / sd,
                 'excludes zero': not (lo < 0 < hi)})
spectator = pd.DataFrame(rows)
print(f"Trend-adjusted contrasts, {len(spectator)} features")
print(f"intervals from bootstrapping matches within season, which describes sampling "
      f"of matches and not assignment of treatment")
print(f"\nintervals excluding zero: {int(spectator['excludes zero'].sum())} "
      f"of {len(spectator)}")
print(f"median absolute effect: "
      f"{spectator['effect (% of sd)'].abs().median():.1f}% of a standard deviation")
print(f"naive comparison would call sizeable (>15% of sd): "
      f"{int((100 * spectator['naive difference'].abs() / panel[SPEC_FEATURES].std().values > 15).sum())}")
spectator.round(3)
```

```python
fig = plt.figure(figsize=(17, 9))
gs = fig.add_gridspec(2, 3, hspace=0.40, wspace=0.28)

# (a) the trend problem, drawn
for k, f in enumerate(['total_passes', 'passes_per_possession', 'strength_gini']):
    ax = fig.add_subplot(gs[0, k])
    m = panel.groupby('t')[f].agg(['mean', 'sem'])
    ctrl = m.drop(index=2)
    slope, intercept = np.polyfit(ctrl.index.values.astype(float), ctrl['mean'].values, 1)
    xs = np.linspace(-0.3, 5.3, 40)
    ax.plot(xs, intercept + slope * xs, color=MIST, lw=2, ls='--',
            label='trend through crowd seasons')
    for t_, r in m.iterrows():
        col = CORAL if t_ == 2 else TEAL
        ax.errorbar(t_, r['mean'], yerr=1.96 * r['sem'], fmt='o', color=col,
                    markersize=11, capsize=4, lw=2)
    ax.set_xticks(list(SEASON_POS.values()))
    ax.set_xticklabels(list(SEASON_POS.keys()), fontsize=8.4, rotation=25)
    ax.set_ylabel(f.replace('_', ' '))
    ax.set_title(f'({chr(97+k)}) {f.replace("_", " ")}', fontsize=10.8)
    if k == 0:
        ax.legend(fontsize=8)

# (d) all effects with intervals
ax = fig.add_subplot(gs[1, :2])
s = spectator.sort_values('effect (% of sd)')
y = np.arange(len(s))
cols = [CORAL if e else MIST for e in s['excludes zero']]
ax.hlines(y, s['CI low (% of sd)'], s['CI high (% of sd)'], color=MIST, lw=3.4)
ax.scatter(s['effect (% of sd)'], y, s=80, color=cols, edgecolor=SLATE,
           linewidth=0.8, zorder=3)
ax.axvline(0, color=SLATE, lw=1.5)
ax.set_yticks(y)
ax.set_yticklabels([f.replace('_', ' ') for f in s.feature], fontsize=8.6)
ax.set_xlabel('Trend-adjusted contrast (% of a standard deviation)')
ax.set_title('(d) Closed-door season against the trend, with match-level intervals',
             fontsize=11)

# (e) naive against adjusted, showing what the adjustment removes
ax = fig.add_subplot(gs[1, 2])
sd_vec = panel[SPEC_FEATURES].std().to_numpy()
ax.scatter(100 * spectator['naive difference'].to_numpy() / sd_vec,
           spectator['effect (% of sd)'], s=75, color=TEAL, edgecolor=SLATE,
           linewidth=0.8)
lim = 60
ax.plot([-lim, lim], [-lim, lim], color=MIST, ls='--', lw=1.4)
ax.axhline(0, color=SLATE, lw=1); ax.axvline(0, color=SLATE, lw=1)
ax.set_xlabel('Naive before and after (% of sd)')
ax.set_ylabel('After trend adjustment (% of sd)')
ax.set_title('(e) What the trend absorbs', fontsize=10.8)

fig.suptitle('Figure 8. The treated season sits on the trend, not off it',
             fontsize=14, color=SLATE, fontweight='bold', y=0.98)
plt.show()
```

<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">What the Design Could Have Detected</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">A null without a power analysis is an absence of information</div></div>

```python
def season_permutation_p(frame, feature):
    """Permutation p from reassigning which season was treated."""
    obs = abs(trend_adjusted_effect(frame, feature, 2))
    null = [abs(trend_adjusted_effect(frame, feature, t))
            for t in sorted(frame.t.unique())]
    return (1 + sum(n >= obs for n in null)) / (1 + len(null))


print("The permutation distribution has one value per season")
print(f"  seasons available      : {panel.t.nunique()}")
print(f"  null distribution size : {panel.t.nunique()}")
print(f"  observed is one of them, so the smallest attainable p is "
      f"{2 / (1 + panel.t.nunique()):.2f}\n")

power_rows = []
base_f = 'density'
sd_f = panel[base_f].std()
for delta in [0.0, 0.25, 0.5, 1.0, 2.0]:
    detected = 0
    recovered = []
    REPS = 60
    for r in range(REPS):
        shifted = panel.assign(**{base_f: panel[base_f] + panel.closed * delta * sd_f})
        est = trend_adjusted_effect(shifted, base_f, 2)
        recovered.append(est / sd_f)
        if season_permutation_p(shifted, base_f) < 0.25:
            detected += 1
    power_rows.append({'injected shift (sd)': delta,
                       'recovered (sd)': float(np.mean(recovered)),
                       'detected at p < 0.25 (%)': 100 * detected / REPS})
power = pd.DataFrame(power_rows)
print("Injecting a known season-level shift and attempting to recover it")
print(power.round(3).to_string(index=False))
print("\nThe estimator recovers the injected shift almost exactly.")
print("The test detects none of them, including a two standard deviation effect.")
```

<div style="background:#FDF2EC;border-left:4px solid #E8694A;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#E8694A;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Result</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">The estimator works and the test cannot</h4>
Injecting a known season-level shift and re-running the pipeline recovers it accurately, since a two
standard deviation shift comes back as approximately two. The estimator is doing its job.
<br><br>
The test detects none of them. Not the two standard deviation shift, not any smaller one. With
four seasons the permutation distribution has four values and the observed estimate is one of
them, so the p-value is bounded below at 0.4 regardless of how large the effect is.
<br><br>
Those two facts are separate and it takes a power analysis to separate them. Reporting the Section 17
null without this would imply that no effect was found when the truth is that no effect could
have been found. The estimates in Section 17 are worth reporting as descriptive contrasts; the
absence of significance in them carries no information at all.
</div>

<div id="sec20" style="border-left:5px solid #E8694A;padding:16px 0 16px 20px;
margin:30px 0 14px;font-family:Helvetica,Arial,sans-serif;">
<span style="color:#E8694A;font-size:12.5px;letter-spacing:2.6px;font-weight:bold;">SECTION 20</span>
<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Predictive Extension</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">Two questions that look identical and are not</div></div>


Asking whether passing structure predicts results turns out to be two different questions with
opposite answers, and the difference is when the features are measured.

Using network features from the match being predicted asks whether structure and outcome go
together. Using features from a club's earlier matches asks whether structure forecasts. The
first is an association and the second is prediction, and conflating them is the most common
way this kind of analysis overstates itself.


```python
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import GroupKFold, cross_val_predict
from sklearn.metrics import roc_auc_score, log_loss, brier_score_loss

tmp = team_match.copy()
tmp['win'] = (tmp.goals > tmp.goals_against).astype(int)
tmp['pts'] = np.where(tmp.goals > tmp.goals_against, 3,
                      np.where(tmp.goals == tmp.goals_against, 1, 0))
tot = tmp.groupby('team').pts.transform('sum')
cnt = tmp.groupby('team').pts.transform('size')
tmp['strength'] = (tot - tmp.pts) / (cnt - 1)      # leave one out, never self-predicting
tmp['opp_strength'] = tmp.opponent.map(tmp.groupby('team').pts.mean())

CONTEXT = ['is_home', 'strength', 'opp_strength']
NETFEAT = ['density', 'weighted_reciprocity', 'strength_gini', 'pass_entropy_norm',
           'clustering', 'total_passes', 'passes_per_possession', 'avg_path_length',
           'progression_mean', 'final_third_share', 'zone_entropy_norm', 'width_iqr']

y_w = tmp.win.to_numpy()
grp = tmp.match_id.to_numpy()      # both sides of a match stay on one side of the split
explain_rows = []
for name, cols in [('match context only', CONTEXT),
                   ('context plus same-match network', CONTEXT + NETFEAT),
                   ('same-match network only', NETFEAT)]:
    X = tmp[cols].astype(float).fillna(tmp[cols].astype(float).median()).to_numpy()
    for mn, md in [('logistic', make_pipeline(StandardScaler(),
                                              LogisticRegression(max_iter=3000))),
                   ('gradient boosting',
                    HistGradientBoostingClassifier(random_state=SEED, max_iter=250))]:
        p = cross_val_predict(md, X, y_w, cv=GroupKFold(5), groups=grp,
                              method='predict_proba')[:, 1]
        explain_rows.append({'features': name, 'model': mn,
                             'AUC': roc_auc_score(y_w, p),
                             'log loss': log_loss(y_w, p),
                             'Brier': brier_score_loss(y_w, p)})
explanatory = pd.DataFrame(explain_rows)
print("A. EXPLANATORY, features measured in the match being predicted")
print(f"   base rate {y_w.mean():.3f}, {len(y_w)} team-matches")
explanatory.round(4)
```

```python
fc = team_match.copy()
fc['date'] = pd.to_datetime(fc.match_date)
fc = fc.sort_values('date').reset_index(drop=True)
fc['win'] = (fc.goals > fc.goals_against).astype(int)
fc['pts'] = np.where(fc.goals > fc.goals_against, 3,
                     np.where(fc.goals == fc.goals_against, 1, 0))

FORECAST_NET = ['density', 'weighted_reciprocity', 'strength_gini', 'total_passes',
                'passes_per_possession', 'progression_mean', 'final_third_share',
                'zone_entropy_norm']
# expanding mean over a club's earlier matches only, shifted so the current match is excluded
for c in FORECAST_NET + ['pts']:
    fc['prior_' + c] = fc.groupby('team')[c].transform(
        lambda s: s.shift().expanding().mean())
fc['played_so_far'] = fc.groupby('team').cumcount()

opp_cols = ['prior_' + c for c in FORECAST_NET + ['pts']]
opp_lookup = fc[['match_id', 'team'] + opp_cols].rename(
    columns={'team': 'opponent', **{c: c.replace('prior_', 'oppprior_') for c in opp_cols}})
fc = fc.merge(opp_lookup, on=['match_id', 'opponent'], how='left')

usable = fc[fc.played_so_far >= 8].dropna(
    subset=['prior_pts', 'oppprior_pts']).copy()
cutoff = usable.date.quantile(0.7)
train, test = usable[usable.date <= cutoff], usable[usable.date > cutoff]

FORM = ['is_home', 'prior_pts', 'oppprior_pts']
FULL = FORM + ['prior_' + c for c in FORECAST_NET] + \
       ['oppprior_' + c for c in FORECAST_NET]

forecast_rows = []
for name, cols in [('form only', FORM), ('form plus prior networks', FULL)]:
    Xtr = train[cols].astype(float).to_numpy()
    Xte = test[cols].astype(float).to_numpy()
    for mn, md in [('logistic', make_pipeline(StandardScaler(),
                                              LogisticRegression(max_iter=3000))),
                   ('gradient boosting',
                    HistGradientBoostingClassifier(random_state=SEED, max_iter=200))]:
        md.fit(Xtr, train.win)
        p = md.predict_proba(Xte)[:, 1]
        forecast_rows.append({'features': name, 'model': mn,
                              'AUC': roc_auc_score(test.win, p),
                              'log loss': log_loss(test.win, p),
                              'Brier': brier_score_loss(test.win, p)})
forecasting = pd.DataFrame(forecast_rows)
print("B. FORECASTING, features from each club's earlier matches only")
print(f"   {len(usable)} rows after an 8-match warm-up")
print(f"   train to {cutoff.date()} ({len(train)}), test after ({len(test)})")
forecasting.round(4)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 5.2))

ax = axes[0]
e = explanatory.groupby('features').AUC.max().reindex(
    ['match context only', 'same-match network only',
     'context plus same-match network'])
ax.barh(range(len(e)), e.values, color=[SLATE, SAND, CORAL],
        edgecolor=SLATE, linewidth=0.7)
ax.axvline(0.5, color=MIST, ls=':', lw=1.6)
ax.set_yticks(range(len(e)))
ax.set_yticklabels([i.replace(' ', '\n') for i in e.index], fontsize=8.2)
ax.set_xlim(0.45, 0.9); ax.set_xlabel('AUC, best model')
ax.set_title('(a) Explanatory, same-match features', fontsize=10.8)

ax = axes[1]
f_ = forecasting.groupby('features').AUC.max().reindex(
    ['form only', 'form plus prior networks'])
ax.barh(range(len(f_)), f_.values, color=[SLATE, CORAL], edgecolor=SLATE, linewidth=0.7)
ax.axvline(0.5, color=MIST, ls=':', lw=1.6)
ax.set_yticks(range(len(f_)))
ax.set_yticklabels([i.replace(' ', '\n') for i in f_.index], fontsize=8.2)
ax.set_xlim(0.45, 0.9); ax.set_xlabel('AUC on a held-out later period')
ax.set_title('(b) Forecasting, prior-match features', fontsize=10.8)

ax = axes[2]
gain_e = (explanatory[explanatory.features == 'context plus same-match network'].AUC.max()
          - explanatory[explanatory.features == 'match context only'].AUC.max())
gain_f = (forecasting[forecasting.features == 'form plus prior networks'].AUC.max()
          - forecasting[forecasting.features == 'form only'].AUC.max())
ax.bar([0, 1], [gain_e, gain_f], color=[CORAL, TEAL], edgecolor=SLATE, linewidth=0.8,
       width=0.58)
ax.axhline(0, color=SLATE, lw=1.4)
ax.set_xticks([0, 1])
ax.set_xticklabels(['same match\n(explanatory)', 'prior matches\n(forecasting)'],
                   fontsize=9)
ax.set_ylabel('AUC gained by adding network features')
for x, v in zip([0, 1], [gain_e, gain_f]):
    ax.text(x, v + (0.002 if v > 0 else -0.004), f'{v:+.4f}', ha='center',
            fontsize=10, color=SLATE, fontweight='bold')
ax.set_title('(c) The same features, opposite verdicts', fontsize=10.8)

fig.suptitle('Figure 9. Explanation and forecasting are not the same question',
             fontsize=14, color=SLATE, fontweight='bold', y=1.02)
plt.tight_layout(); plt.show()

print(f"Network features add {gain_e:+.4f} AUC when measured in the same match")
print(f"Network features add {gain_f:+.4f} AUC when measured beforehand")
```

With features from the match being predicted, adding network structure raises AUC by 0.028. With features from each club's earlier matches only, adding the historical network profile makes
forecasting 0.013 AUC worse, with log loss and Brier score
both deteriorating.
<br><br>
The gap is not a modelling artefact. It follows directly from the variance decomposition, where 74.3%
of network variation sits at match level, so a club's historical average carries little
information about how it will play next week. There is not much signal to forecast with.
<br><br>
The explanatory result also has a reading that is easy to miss. A team completing 500 passes and
dominating territory is partly a team that is winning, because sides behave differently when
ahead. Some of that 0.028 AUC gain is the outcome leaking into the features rather than the
features anticipating the outcome, and no cross-validation scheme can remove it when the
features are recorded during the match.
<br><br>
Gradient boosting underperforms logistic regression in the forecasting setup, 0.652 against
0.708, which is what a flexible model does to a weak signal.

<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Discussion</h2><div style="color:#5A7480;font-size:12.5px;margin-top:7px;font-style:italic;">What a passing network is evidence about</div></div>


The central result is that a passing network is mostly evidence about a match. Averaged over
22 features, 74.3% of the variance in network structure sits at match level, 23.4% at club
level and 2.2% at league level. Every other finding in this project follows from that split.

It explains why a classifier given one network can barely name its league, reaching 0.378
against a chance level of 0.250. It explains why network features measured during a match
appear to predict its result and the same features measured beforehand do not. And it explains
why five of twelve associations between network structure and goal difference reverse sign once
team quality is controlled, because what looked like structure predicting outcomes was mostly
good teams doing both.

The finding runs against a common framing in this literature. Work on Guardiola's Barcelona
identified network metrics that distinguished that side from its opponents, and a substantial
body of applied analysis treats the passing network as a stable signature of how a team plays.
This project does not contradict those results. A side as distinctive as that Barcelona team
may well separate from its opponents on network metrics, and the present analysis finds clubs
do differ, accounting for almost a quarter of variance. What it adds is the denominator. Against
the match-to-match variation that any team produces, club identity is a modest signal, and a
single-match network is a noisy draw rather than a portrait.

The one exception is individual. A player's involvement share repeats across matches with the
player mean explaining 46.6% of its variance, roughly twice what any club mean explains for a
team-level feature. Who holds a passing network together is a property of the person. How the
team's graph looks overall is largely a property of the afternoon.

Where leagues do differ, they differ spatially. Final-third share and mean pass origin carry
league shares of 12.7% and 10.6%, while density, reciprocity and centralization carry almost
none. The Premier League plays four metres higher up the pitch than Ligue 1. Its passing graphs
are wired the same way.

color:#2F4750;font-size:14px;"><div style="color:#2AA396;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Interpretation</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">The methodological thread</h4>
Four results in this project changed after being checked, and none of the checks was expensive.
<br><br>
Controlling for team quality reversed five of twelve outcome associations. Adjusting for a
secular trend removed four apparently sizeable spectator effects. Varying the network definition
destroyed one metric entirely, with betweenness centralization correlating at -0.04 with its own
thresholded version. And a power analysis showed the spectator test could not reject at any
effect size, which turned an uninformative null into a stated limit.
<br><br>
Three data-integrity failures were caught the same way, by comparing a computed quantity against
something known independently. Two clubs with zero passes, two clubs split into four identities,
and 128 matches with a goal missing because own goals are recorded separately. None raised an
error, and the league-table reconstruction in Section 5.1 is the check that would have caught
all three at once.
</div>

<h2 style="color:#22414E;margin:6px 0 0;font-size:24px;font-weight:600;">Limitations</h2></div>

<div style="background:#F1F5F6;border-left:4px solid #6E8A95;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#6E8A95;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Limitation</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">The samples</h4>
The cross section is one season. Everything said about clubs describes those 80 clubs in
2015/16, and a club's structure in a single season confounds squad, manager and circumstance.
The four leagues are the four with complete coverage in the archive, not a sample of world
football, and the Bundesliga is absent because its release covers only one club's matches.
<br><br>
Ligue 1 is missing three of 380 fixtures, which is why Paris Saint-Germain reconstructs at 93
points rather than 96.
<br><br>
The panel is four seasons of one competition with a two-season gap, one incomplete season, and
treatment assigned at season level. Section 19 quantifies what that costs.
</div>

<div style="background:#F1F5F6;border-left:4px solid #6E8A95;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#6E8A95;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Limitation</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">The measurements</h4>
A passing network built from event data records who passed to whom and where from. It does not
record off-ball movement, defensive shape, or what the players were instructed to do. Node
positions are average pass origins, which is a proxy for position and not a measurement of it.
<br><br>
Completed passes only means the network omits attempted play that failed, and a side whose
ambitious passes are intercepted looks structurally similar to one that never attempts them.
Incomplete passes are counted separately but do not enter the topology.
<br><br>
Sequence is discarded entirely. Two teams with identical adjacency matrices may have arrived
there through completely different patterns of play, and no metric here can separate them. The
animation in Section 16 shows what that loses.
</div>

<div style="background:#F1F5F6;border-left:4px solid #6E8A95;border-radius:3px;
padding:20px 24px;margin:16px 0;font-family:Helvetica,Arial,sans-serif;line-height:1.8;
color:#2F4750;font-size:14px;"><div style="color:#6E8A95;font-size:11px;letter-spacing:2.2px;text-transform:uppercase;margin-bottom:8px;font-weight:bold;">Limitation</div><h4 style="color:#22414E;margin:0 0 10px;font-size:15.5px;">What is not claimed</h4>
Nothing in this project is a causal estimate. The variance decomposition is descriptive, the
outcome models are associational with quality controlled rather than randomised, and the
spectator analysis is explicitly a bounded case study.
<br><br>
The forecasting result shows network features do not improve out-of-sample prediction on this
sample with these features and this horizon. It does not show that no network representation
could.
</div>

deviations.</p>

<p style="margin:0 0 15px;"><strong>What the data do not support.</strong> National leagues have
no meaningful topological signature. A classifier given one network reaches 0.378 balanced
accuracy at naming its league against 0.250 chance, and clustering 80 clubs on their passing
features recovers the league labels at an adjusted Rand of -0.005. Network structure does not
predict match outcomes once team quality is controlled, with five of twelve associations
reversing sign. And structure measured beforehand makes forecasting slightly worse rather than
better.</p>

<p style="margin:0 0 15px;"><strong>What could not be tested.</strong> The spectator analysis
returns a null that carries no information. The closed-door season sits on the trend running
through the seasons around it rather than off it, and with treatment assigned at season level
across four seasons the permutation test cannot reject at any effect size. The estimator
recovers injected shifts of two standard deviations accurately and the test detects none of
them. That is a property of the design, not of the football.</p>

<p style="margin:0;"><strong>What the pipeline can be trusted on.</strong> Every goal in this
corpus was reconstructed from individual shot and own-goal events. Aggregated into season
points, those reconstructions reproduce the official 2015/16 league tables exactly for nine
clubs across three leagues on seven quantities each, including Leicester City at 81 points from
23 wins, 12 draws and 3 defeats. That is the check that tests parsing, own-goal handling, club
identity and home assignment simultaneously, and it is the reason the rest of the analysis can
be read as being about football rather than about a data pipeline.</p>
